# 1. Reisezeit

Basierend auf Transportmethode braucht es mehr oder weniger Zeit um eine Strecke zurückzulegen.
In diesem Programm soll der Benutzer die Transportmethode und die Strecke eingeben.
Als Ausgabe soll die Reisezeit in Stunden und Minuten ausgegeben werden.

| Transportmethode | Geschwindigkeit in km/h |
|------------------|-------------------------|
| Gehen (walk)             | 5                     |
| Fahrrad (bike)              | 15                    |
| Auto (car)             | 50                    |
| Zug (train)              | 70                    |


````python
transportmethode = input("Wie kommst du vorwärts (walk, bike, car or train)")
strecke = float(input("Wie lang ist die Strecke in km?"))

if transportmethode == "walk":
    geschwindigkeit = 5
elif transportmethode == "bike":
    geschwindigkeit = 15
elif transportmethode == "car":
    geschwindigkeit = 50
elif transportmethode == "train":
    geschwindigkeit = 70
else:
    print("Ungültige Transportmethode")

stunden = strecke / geschwindigkeit
minuten = (stunden - int(stunden)) * 60
sekunden = (minuten - int(minuten)) * 60
print("Du brauchst",int(stunden),"Stunden,",int(minuten),"Minuten und",int(sekunden),"Sekunden")
````

# 3. Taschenrechner

Schreibe einen Taschenrechner, mit folgenden Eigenschaften:

- Er fragt nach der ersten Zahl
- Dann nach dem Operator (also +, -, /, und *)
- Dann nach der zweiten Zahl
- Anschliessend berechnet er das Resultat


````python
zahl1 = float(input("Erste Zahl?"))
operator = input("Operator: +, -, * oder /")
zahl2 = float(input("Zweite Zahl"))

if operator == "+":
    print(zahl1 + zahl2)
elif operator == "-":
    print(zahl1 - zahl2)
elif operator == "*":
    print(zahl1 * zahl2)
elif operator == "/":
    print(zahl1 / zahl2)
else:
    print("Ungültiger Operator")
````
