### 2.3 Datentyp umwandeln

Manchmal wollen wir einen Wert von einem Datentyp in einen anderen umwandeln. Das nennt man Konvertierung. Zum Beispiel können wir eine Zahl in einen Text verwandeln, wenn wir sie mit anderen Texten verbinden möchten.

````python
alter = 48
#"Hallo, ich bin " + alter + " Jahre alt"
````

Nein, ein Zusammenfügen von `str` und `int` ist nicht möglich. Deshalb müssen wir die Zahl zuerst in einen Text umwandeln.

````python
# Umwandlung von int zu str
alter_as_str = str(alter)

# Welcher Typ hat die neue Variable?
type(alter_as_str)
````

Nun können wir das konvertierte Alter in den Text einfügen:

````python
"Hallo, ich bin " + alter_as_str + " Jahre alt"
````

Wir müssen dazu übrigens nicht zwingend eine neue Variable erstellen, sondern das Alter direkt konvertieren:

````python
"Hallo, ich bin " + str(alter) + " Jahre alt"
````

````python
some_float
````

## 3. Verwendung von Funktionen

Funktionen helfen uns, Aufgaben zu wiederholen, ohne immer denselben Code neu zu schreiben. Zum Beispiel gibt uns `type()` Auskunft über den Datentyp eines Werts.

````python
type
````

Wenn wir eine Funktion mit der falschen Anzahl an Argumenten aufrufen, erhalten wir einen Fehler. Das ist ein guter Grund, die Funktionssyntax genau zu prüfen.

````python
type()
````

Die Funktion `type` erwartet also ein oder drei Argumente. Das zeigt, dass Funktionen je nach Verwendung unterschiedliche Anforderungen haben.

Die meisten Funktionen wie `type`, `str`, `float` und `int` berechnen ein Ergebnis und geben es zurück. Das Ergebnis können wir speichern oder direkt weiterverwenden.

````python
my_number = 3.15
int(my_number)   #Hier wird der neue Wert nicht gespeichert
print(my_number)

my_number = 3.15
my_number = int(my_number) # Hier wird der neue Wert der Variabel zugewiesen
print(my_number)

my_number = 3.15
my_int_number = int(my_number) # Hier wird der neue Wert einer neuen Variabel zugewiesen
print(my_number)
print(my_int_number)
````

Das Ergebnis können wir direkt weiterverwenden:

````python
3 + int(3.15)
````

Wir können das Ergebnis einer Funktion auch direkt in eine andere Funktion geben:

````python
str(int(3.15))

my_value = 3.15
my_int = int(my_value)
my_string = str(my_int)
my_string
````

Auch innerhalb eines Funktionsarguments können wir Rechnungen ausführen. Python löst sie von innen nach aussen auf.

### 4.3 Weiterverarbeitung von Strings als Zahlen

Wichtig: `input()` liefert immer einen String. Wenn wir eine Zahl haben wollen, müssen wir sie umwandeln.

````python
# Abfrage mit input
answer = input('In welchem Jahr bist du geboren? ')

# Wir konvertieren den str zu einer ganzen Zahl (int)
birth_year = int(answer)

# Nun können wir damit rechnen:
age = 2024 - birth_year

# Und das Resultat zurückgeben
print('Du wirst im Jahr 2024', age, 'Jahre alt.')
````

Wenn wir diese Umwandlung nicht machen, erhalten wir einen Fehler. Das ist ein typischer Anfängerfehler.

````python
# Abfrage mit input
answer = input('In welchem Jahr bist du geboren? ')

# Wir rechnen direkt mit der Antwort, ohne Umrechnung --> gibt Fehler!
age = 2023 - answer

# Rückgabe des Resultats
print('Du wirst im Jahr 2023', age, 'Jahre alt.')
````
