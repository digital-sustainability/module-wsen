Loesungen zu den Aufgaben zu Comprehensions und String-Formatierung.

## A) Comprehensions

### Aufgabe 1

```python
namen = ["anna", "ben", "cora"]
gross = [n.upper() for n in namen]
print(gross)
```

### Aufgabe 2

```python
gerade = [z for z in range(21) if z % 2 == 0]
print(gerade)
```

### Aufgabe 3

```python
woerter = ["python", "oop", "datenbank", "api"]
laengen = {wort: len(wort) for wort in woerter}
print(laengen)
```

### Aufgabe 4

```python
werte = [2, 4, 6]
verdoppelt = [w * 2 for w in werte]
print(verdoppelt)
```

## B) String-Formatierung

### Aufgabe 5

```python
name = "Lea"
punkte = 14
print(f"{name} hat {punkte} Punkte erreicht.")
```

### Aufgabe 6

```python
pi = 3.1415926
print(f"{pi:.3f}")
```

### Aufgabe 7

```python
anteil = 0.256
print(f"{anteil:.1%}")
```

### Aufgabe 8

```python
text = "Hallo {name}, Ihr Termin ist am {datum}."
print(text.format(name="Nora", datum="05.12.2026"))
```

## C) Transfer

### Aufgabe 9

```python
preise = [4.5, 19, 0.99, 12.345]
texte = [f"Artikel {i + 1}: {preis:.2f} CHF" for i, preis in enumerate(preise)]
for t in texte:
    print(t)
```

### Aufgabe 10

```python
noten = {"Ava": 5.0, "Liam": 3.5, "Mia": 4.0}
status = {
    name: ("bestanden" if note >= 4.0 else "nicht bestanden")
    for name, note in noten.items()
}
print(status)
```
