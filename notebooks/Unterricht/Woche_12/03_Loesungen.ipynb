## A) Comprehensions

### Aufgabe 1
Die List-Comprehension verwendet denselben Ausdruck wie die Schleife, nur kompakter: `n.upper()` wird für jedes Element aus `namen` berechnet.


```python
namen = ["anna", "ben", "cora"]
gross = [n.upper() for n in namen]
print(gross)
```

### Aufgabe 2
Die Bedingung `z % 2 == 0` lässt nur gerade Zahlen durch, deshalb bleibt die Ausgabe klein und verständlich.


```python
gerade = [z for z in range(21) if z % 2 == 0]
print(gerade)
```

### Aufgabe 3
Ein Dictionary kann direkt aus einem Iterable erzeugt werden, indem man Schlüssel und Werte im Ausdruck angibt.


```python
woerter = ["python", "oop", "datenbank", "api"]
laengen = {wort: len(wort) for wort in woerter}
print(laengen)
```

### Aufgabe 4
Die Comprehension braucht den Ausdruck für jedes Element direkt vor dem `for`; der Fehler ist deshalb nur ein Syntaxproblem.


```python
werte = [2, 4, 6]
verdoppelt = [w * 2 for w in werte]
print(verdoppelt)
```

## B) String-Formatierung

### Aufgabe 5
Mit f-Strings lassen sich Variablen direkt in einen Text einfügen, ohne zusätzliche Zeichenkettenverkettung.


```python
name = "Lea"
punkte = 14
print(f"{name} hat {punkte} Punkte erreicht.")
```

### Aufgabe 6
`:.3f` formatiert die Zahl auf drei Nachkommastellen und rundet sie passend.


```python
pi = 3.1415926
print(f"{pi:.3f}")
```

### Aufgabe 7
`:.1%` wandelt die Zahl in Prozentdarstellung um und formatiert sie auf eine Nachkommastelle.


```python
anteil = 0.256
print(f"{anteil:.1%}")
```

### Aufgabe 8
`.format(...)` ersetzt benannte Platzhalter sauber und ist besonders nützlich für Vorlagen.


```python
text = "Hallo {name}, Ihr Termin ist am {datum}."
print(text.format(name="Nora", datum="05.12.2026"))
```

## C) Transfer

### Aufgabe 9
Die Liste wird elementweise aufgebaut; `enumerate` liefert die laufende Nummer und den jeweiligen Preis.


```python
preise = [4.5, 19, 0.99, 12.345]
texte = [f"Artikel {i + 1}: {preis:.2f} CHF" for i, preis in enumerate(preise)]
for t in texte:
    print(t)
```

### Aufgabe 10
Die Comprehension wertet jedes Paar `name, note` aus und erzeugt so ein Dictionary mit dem passenden Status.


```python
noten = {"Ava": 5.0, "Liam": 3.5, "Mia": 4.0}
status = {
    name: ("bestanden" if note >= 4.0 else "nicht bestanden")
    for name, note in noten.items()
}
print(status)
```


## Musterlösung: Formatierte Ausgabe, JSON und CSV


```python
import csv, io, json
rechnung = {"rechnungsnummer":"R-12", "datum":"2024-11-25", "kunde":{"name":"Alpenblick AG"}, "positionen":[{"artikel":{"bezeichnung":"Beratung"},"menge":2,"einzelpreis":150.0,"mwst_satz":7.7,"betrag":300.0}], "summen":{"mwst":{"7.7":23.1},"gesamtbetrag":323.1}}
steuer_77 = [p for p in rechnung["positionen"] if p.get("mwst_satz") == 7.7]
zeilen = [f"{p['artikel'].get('bezeichnung','(ohne Bezeichnung)')} | {p['menge']} x CHF {p['einzelpreis']:.2f} | CHF {p['betrag']:.2f}" for p in rechnung["positionen"]]
text = f"RECHNUNG {rechnung['rechnungsnummer']}\nKunde: {rechnung['kunde'].get('name','unbekannt')}\n" + "\n".join(zeilen) + f"\nMWST 7.7%: CHF {sum(p['betrag']*p['mwst_satz']/100 for p in steuer_77):.2f}\nTOTAL: CHF {rechnung['summen']['gesamtbetrag']:.2f}"
print(text); assert json.loads(json.dumps(rechnung)) == rechnung
buf=io.StringIO(); w=csv.writer(buf); w.writerow(["Rechnung","Position","Betrag"]); w.writerows([[rechnung['rechnungsnummer'],i,f"{p['betrag']:.2f}"] for i,p in enumerate(rechnung['positionen'],1)]); print(buf.getvalue())
```
