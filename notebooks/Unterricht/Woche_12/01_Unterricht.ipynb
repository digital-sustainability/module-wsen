## 1. Comprehensions

### 1.1 Idee und Aufbau
Comprehensions sind eine kompaktere Schreibweise für eine Schleife, wenn du jedes Element aus einer Liste oder einer anderen Datenstruktur verarbeiten willst.
Am Anfang ist es hilfreich, sie als kurze Variante einer Schleife zu sehen: statt viele Zeilen für den gleichen Schritt zu schreiben, beschreibst du den Schritt in einem Ausdruck.

Eine Comprehension hat meist die Form:

`[ausdruck for element in quelle if bedingung]`

- `ausdruck`: Was passiert mit jedem Element?
- `quelle`: Aus welcher Datenstruktur kommen die Werte?
- `if bedingung` (optional): Welche Elemente bleiben erhalten?

Beispiel: Wir wollen aus jeder Zahl ihr Quadrat machen.


```python
zahlen = [1, 2, 3, 4, 5]

# 1) Schleife (langsam, aber leicht nachvollziehbar)
quadrate_loop = []
for z in zahlen:
    quadrate_loop.append(z * z)

# 2) Comprehension (kurz und kompakt)
quadrate_comp = [z * z for z in zahlen]

print(quadrate_loop)
print(quadrate_comp)
```

Mini-Check: Was ist der Unterschied zwischen `quadrate_loop` und `quadrate_comp`?
Warum ist die Comprehension für dieses Beispiel kürzer, aber trotzdem leicht lesbar?


### 1.2 Filtern mit `if`

```python
werte = [-3, -1, 0, 2, 7]
positive = [w for w in werte if w > 0]
gerade_positive = [w for w in werte if w > 0 and w % 2 == 0]

print(positive)
print(gerade_positive)
```

### 1.3 Ausdruck mit `if ... else` (Werte umschreiben)

```python
noten = [5.0, 3.5, 4.0, 2.8]
status = ["bestanden" if n >= 4.0 else "nicht bestanden" for n in noten]
print(status)
```

### 1.4 Set- und Dictionary-Comprehensions

```python
namen = ["anna", "ben", "anna", "cora", "ben"]
eindeutige_namen = {n.upper() for n in namen}
print(eindeutige_namen)
```

```python
preise = {"apfel": 2.5, "birne": 3.0, "banane": 1.8}
rabatt_preise = {name: round(preis * 0.9, 2) for name, preis in preise.items()}
teure_artikel = {name: preis for name, preis in preise.items() if preis >= 2.0}

print(rabatt_preise)
print(teure_artikel)
```

### 1.5 Verschachtelte Comprehensions (mit Vorsicht)
Verschachtelte Comprehensions sind möglich, aber oft schwieriger zu lesen.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Alle Werte in eine flache Liste
flach = [wert for zeile in matrix for wert in zeile]
print(flach)
```

### 1.6 Wann nicht?
Comprehensions passen gut, wenn jedes Element unabhängig verarbeitet wird.
Wenn Zwischenzustände wichtig sind, ist eine Schleife meist klarer.

```python
# Kein guter Comprehension-Fall: laufende Summe
running_sum = 0
result = []
for i in [1, 2, 3, 4]:
    running_sum += i
    result.append(running_sum)

print(result)
```

## Anwendung im Projekt: Rechnungspositionen mit Comprehensions aufbereiten

Aus dem gemeinsamen Rechnungs-Dictionary bauen wir mit Comprehensions Listen für die Ausgabe: eine Zeile pro Position und eine Zeile pro MWST-Satz.

```python
rechnung = {"rechnungsnummer":"R-12", "datum":"2024-11-25", "kunde":{"name":"Alpenblick AG"}, "positionen":[{"artikel":{"bezeichnung":"Beratung"},"menge":2,"einzelpreis":150.0,"mwst_satz":7.7,"betrag":300.0}], "summen":{"zwischensumme":300.0,"mwst":{"7.7":23.1},"gesamtbetrag":323.1}}
zeilen = [f"{p['artikel'].get('bezeichnung','(ohne Bezeichnung)'):20} {p['menge']:>3} x CHF {p['einzelpreis']:>8.2f} = CHF {p['betrag']:>8.2f}" for p in rechnung["positionen"]]
steuerzeilen = [f"MWST {satz}%: CHF {betrag:.2f}" for satz, betrag in rechnung["summen"].get("mwst", {}).items()]
print(zeilen)
print(steuerzeilen)
```

## 2. String-Formatierung: Ausgaben lesbar machen


### 2.1 Warum formatieren?
Formatierte Ausgaben sind leichter zu lesen, weil Menschen Zahlen und Texte schneller verstehen, wenn sie klar strukturiert sind.
Statt eine Ausgabe mit vielen Zeichen zu bauen, kann man sie mit einer festen Form darstellen.

Beispiel: `print("Preis:", preis)` ist weniger klar als `print(f"Preis: {preis:.2f} CHF")`.

- Lesbarkeit der Ausgabe verbessern
- Zahlen kontrolliert darstellen (Rundung, Prozent, wissenschaftliche Notation)
- Tabellen und Reports sauber ausrichten


### 2.2 f-Strings: schnelle und moderne Variante

```python
name = "Mila"
punkte = 17
print(f"{name} hat {punkte} Punkte.")
```

Mini-Check: Warum ist `f"{name} hat {punkte} Punkte."` klarer als eine Ausgabe mit mehreren `+`-Zeichen?


### 2.3 Zahlen formatieren
- `:.2f` -> 2 Nachkommastellen
- `:.1%` -> Prozentdarstellung
- `:.2e` -> wissenschaftliche Notation

```python
preis = 12.3456
anteil = 0.3781
klein = 0.0015

print(f"Preis: {preis:.2f} CHF")
print(f"Anteil: {anteil:.1%}")
print(f"Wissenschaftlich: {klein:.2e}")
```

### 2.4 Breite und Ausrichtung
- `<` linksbündig
- `>` rechtsbündig
- `^` zentriert

```python
wort = "hallo"
print(f"|{wort:<10}|")
print(f"|{wort:>10}|")
print(f"|{wort:^10}|")
```

### 2.5 `.format(...)`: nützlich für Vorlagen

```python
template = "Guten Tag {name}, Ihr Kontostand betraegt {betrag:.2f} CHF."
text = template.format(name="Noah", betrag=1532.5)
print(text)
```

```python
mail = """Hallo {name},
deine naechste Lektion startet um {zeit} Uhr.
Freundliche Gruesse,
Dozententeam"""

print(mail.format(name="Lea", zeit="13:15"))
```

### 2.6 Mini-Beispiel: einfache Tabelle

```python
artikel = [
    ("Apfel", 2.5, 4),
    ("Birne", 3.0, 2),
    ("Banane", 1.8, 7)
]

print(f"{'Artikel':<10} {'Preis':>8} {'Menge':>8} {'Total':>10}")
for name, preis, menge in artikel:
    total = preis * menge
    print(f"{name:<10} {preis:>8.2f} {menge:>8d} {total:>10.2f}")
```

## Anwendung im Projekt: Menschen- und maschinenlesbare Ausgabe

Mit den in Comprehensions vorbereiteten Zeilen bauen wir jetzt eine lesbare `f-String`-Ausgabe und exportieren dieselbe Rechnung zusätzlich als JSON und CSV.

```python
import csv, io, json

text = "\n".join([f"RECHNUNG {rechnung['rechnungsnummer']} | {rechnung['datum']}", f"Kunde: {rechnung['kunde'].get('name','unbekannt')}", *zeilen, *steuerzeilen, f"TOTAL: CHF {rechnung['summen']['gesamtbetrag']:.2f}"])
print(text)
json_text = json.dumps(rechnung, ensure_ascii=False)
out = io.StringIO(); writer = csv.DictWriter(out, fieldnames=["rechnungsnummer","positionsnummer","bezeichnung","menge","betrag"]); writer.writeheader()
for i,p in enumerate(rechnung["positionen"], 1): writer.writerow({"rechnungsnummer":rechnung["rechnungsnummer"],"positionsnummer":i,"bezeichnung":p["artikel"].get("bezeichnung",""),"menge":p["menge"],"betrag":f"{p['betrag']:.2f}"})
csv_text = out.getvalue(); assert json.loads(json_text) == rechnung; print(csv_text)
```

## 3. Kombination: Comprehension + formatierte Ausgabe

```python
umsatz = [23.4, 11.0, 7.95, 88.1]
labels = [f"Position {i + 1}: {wert:.2f} CHF" for i, wert in enumerate(umsatz)]

for zeile in labels:
    print(zeile)
```

```python
noten = {"Ava": 5.0, "Liam": 3.5, "Mia": 4.0}
berichte = [
    f"{name:<6} | Note: {note:.1f} | {'bestanden' if note >= 4.0 else 'nicht bestanden'}"
    for name, note in noten.items()
]

for b in berichte:
    print(b)
```
