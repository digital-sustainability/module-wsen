## 1. Lambda-Grundlagen

### 1.1 Lambda statt kurzer `def`
Hier werden drei sehr kurze Funktionen direkt als `lambda` geschrieben.


```python
verdoppeln = lambda x: x * 2
ist_gerade = lambda x: x % 2 == 0
punktzahl = lambda note: note * 20

print(verdoppeln(6), ist_gerade(7), punktzahl(4.5))
```

### 1.2 Sortieren mit `key=lambda`
Die Sortierfunktion gibt genau den Wert zurück, nach dem sortiert werden soll.


```python
resultate = [("Lea", 72), ("Noah", 88), ("Mia", 81), ("Jan", 88)]
sortiert = sorted(resultate, key=lambda eintrag: eintrag[1], reverse=True)
print(sortiert)
```

Hinweis: `eintrag[1]` ist die Punktzahl. `reverse=True` macht absteigende Reihenfolge.
So sieht man gut, dass `key` nur den Vergleichswert liefert und nicht die ganze Sortierlogik.


### 1.3 Filtern mit `lambda`
Hier behalten wir nur Dateien, die auf `.txt` enden.


```python
dateien = ["daten.csv", "notiz.txt", "bild.png", "todo.txt", "main.py"]
nur_txt = list(filter(lambda name: name.endswith('.txt'), dateien))
print(nur_txt)
```


## Musterlösung: Lambda-Sortierung und rekursive Summe


```python
positionen = [{"positionsnummer": 1, "betrag": 300.0}, {"positionsnummer": 2, "betrag": 40.0}]
sortiert = sorted(positionen, key=lambda p: p["betrag"])
def summe_rekursiv(ps):
    if len(ps) == 0: return 0.0
    return ps[0]["betrag"] + summe_rekursiv(ps[1:])
def summe_iterativ(ps):
    return sum(p["betrag"] for p in ps)
for test in [[], [{"betrag": 9.0}], positionen]:
    assert summe_rekursiv(test) == summe_iterativ(test)
print(sortiert)
```

## 2. Rekursion-Basics

### 2.1 Countdown
Der Basisfall stoppt bei `0`, und jeder Schritt macht das Problem kleiner.


```python
def countdown(n):
    if n <= 0:
        print("Start!")
        return
    print(n)
    countdown(n - 1)

countdown(4)
```

### 2.2 Summe rekursiv
Die leere Liste ist der Basisfall. Danach wird immer das erste Element addiert.


```python
def summe_rekursiv(liste):
    if len(liste) == 0:
        return 0
    return liste[0] + summe_rekursiv(liste[1:])

print(summe_rekursiv([3, 5, 2]))
print(summe_rekursiv([]))
```

### 2.3 Fakultät rekursiv
Auch hier ist der Basisfall entscheidend: Bei `0` oder `1` ist das Ergebnis sofort bekannt.


```python
def fakultaet(n):
    if n <= 1:
        return 1
    return n * fakultaet(n - 1)

print(fakultaet(0))
print(fakultaet(5))
```

### 2.4 Wertsuche rekursiv
Die Funktion prüft erst den Basisfall und dann das erste Element der Liste.


```python
def enthaelt_wert(liste, ziel):
    if len(liste) == 0:
        return False
    if liste[0] == ziel:
        return True
    return enthaelt_wert(liste[1:], ziel)

print(enthaelt_wert([10, 20, 30], 20))
print(enthaelt_wert([10, 20, 30], 99))
```

## 3. Transfer

### 3.1 Rekursive Zeichenkette
Hier kombinieren wir den Basisfall mit einer kleinen Problemverkleinerung pro Aufruf.


```python
def anzahl_zeichen(text, zeichen):
    if text == "":
        return 0

    erster_treffer = 1 if text[0] == zeichen else 0
    return erster_treffer + anzahl_zeichen(text[1:], zeichen)

print(anzahl_zeichen("rekursion", "r"))
print(anzahl_zeichen("banana", "a"))
```

Kurzcheck bei Fehlern:
- Basisfall erreicht?
- Wird das Problem pro Aufruf kleiner (`n-1`, `liste[1:]`, `text[1:]`)?
- Stimmen Rückgabewerte (`return`) in allen Fällen?
