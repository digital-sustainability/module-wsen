## 1. Lambda-Grundlagen

### 1.1 Lambda statt kurzer `def`
Jede kurze Funktion lässt sich direkt als `lambda`-Ausdruck schreiben.


```python
verdoppeln = lambda x: x * 2
ist_gerade = lambda x: x % 2 == 0
punktzahl = lambda note: note * 20

print(verdoppeln(6), ist_gerade(7), punktzahl(4.5))
```

### 1.2 Sortieren mit `key=lambda`
Der Schlüssel ist ein Tupel: zuerst die Punktzahl negiert (absteigend), danach der Name (alphabetisch).


```python
resultate = [("Lea", 72), ("Noah", 88), ("Mia", 81), ("Jan", 88)]
sortiert = sorted(resultate, key=lambda eintrag: (-eintrag[1], eintrag[0]))
print(sortiert)
```

### 1.3 Filtern mit `lambda`
Die Lambda-Funktion gibt für Dateinamen mit der Endung `.txt` `True` zurück.


```python
dateien = ["daten.csv", "notiz.txt", "bild.png", "todo.txt", "main.py"]
nur_txt = list(filter(lambda name: name.endswith(".txt"), dateien))
print(nur_txt)
```

## Anwendungsaufgabe: Podcast-Katalog
Zuerst werden die Episoden alphabetisch nach Titel sortiert. Für die längste Episode wird jeweils das erste Element mit dem Ergebnis für den Rest der Liste verglichen; bei gleicher Dauer bleibt das zuerst gefundene Element erhalten.


```python
episoden = [
    {"titel": "Im Wald", "minuten": 28},
    {"titel": "Code im Alltag", "minuten": 42},
    {"titel": "Kurze Pause", "minuten": 19}
]

sortierte_episoden = sorted(episoden, key=lambda episode: episode["titel"])

def laengste_episode(episoden):
    if len(episoden) == 1:
        return episoden[0]

    laengste_im_rest = laengste_episode(episoden[1:])
    if episoden[0]["minuten"] >= laengste_im_rest["minuten"]:
        return episoden[0]
    return laengste_im_rest

print(sortierte_episoden)
print(laengste_episode([{"titel": "Einzelne Episode", "minuten": 12}]))
print(laengste_episode(episoden))
```

## 2. List-Slicing
Die Endposition eines Slices ist ausgeschlossen. So entsprechen `3:7` den Indizes 3 bis 6; `::3` beginnt bei Index 0 und nimmt jeden dritten Wert.


```python
messwerte = [14, 18, 12, 21, 17, 25, 19, 23, 16, 20]

werte_indizes_3_bis_6 = messwerte[3:7]
jeder_dritte_wert = messwerte[::3]
ohne_ersten_und_letzten_zwei = messwerte[2:-2]

print(werte_indizes_3_bis_6)
print(jeder_dritte_wert)
print(ohne_ersten_und_letzten_zwei)
```

## 3. Rekursion

### 3.1 Quersumme
Bei einer einstelligen Zahl ist die Quersumme die Zahl selbst. Sonst addieren wir die letzte Ziffer zur Quersumme des verbleibenden Zahlenteils.


```python
def quersumme(n):
    if n < 10:
        return n
    return n % 10 + quersumme(n // 10)

print(quersumme(583))
print(quersumme(7))
```

### 3.2 Kleinster Wert
Für eine Liste mit einem Element ist dieses der kleinste Wert. In jedem weiteren Schritt vergleichen wir das erste Element mit dem kleinsten Wert im Rest.


```python
def kleinster_wert(liste):
    if len(liste) == 1:
        return liste[0]

    kleinster_im_rest = kleinster_wert(liste[1:])
    if liste[0] < kleinster_im_rest:
        return liste[0]
    return kleinster_im_rest

print(kleinster_wert([8, -3, 5, -11, 4]))
print(kleinster_wert([6]))
```

### 3.3 Palindrom prüfen
Leere und einstellige Texte sind Palindrome. Stimmen die äusseren Zeichen überein, prüfen wir rekursiv den Text dazwischen.


```python
def ist_palindrom(text):
    if len(text) <= 1:
        return True
    if text[0] != text[-1]:
        return False
    return ist_palindrom(text[1:-1])

print(ist_palindrom("lagerregal"))
print(ist_palindrom("python"))
```

## 4. Transfer

### 4.1 Zeichen entfernen
Ist das erste Zeichen das zu entfernende Zeichen, lassen wir es weg. Andernfalls behalten wir es und setzen es vor das rekursive Ergebnis für den Rest.


```python
def entferne_zeichen(text, zeichen):
    if text == "":
        return ""
    if text[0] == zeichen:
        return entferne_zeichen(text[1:], zeichen)
    return text[0] + entferne_zeichen(text[1:], zeichen)

print(entferne_zeichen("banane", "a"))
print(entferne_zeichen("mississippi", "s"))
```
