## 1. Lambda-Funktionen

### 1.1 Warum benutzen wir `lambda`?

Eine normale Funktion schreiben wir mit `def`. Das ist gut fuer laengere Logik.
Eine `lambda` ist praktisch, wenn wir nur einen kurzen Ausdruck direkt uebergeben wollen.


```python
def quadrat(x):
    return x * x

print(quadrat(4))
```

### 1.2 Was ist eine `lambda`-Funktion?

Eine `lambda`-Funktion ist eine kurze, anonyme Funktion.
Sie eignet sich fuer **kleine Berechnungen**, die wir nur direkt an einer Stelle brauchen.

Allgemeine Form:

`lambda parameter: ausdruck`

Die wichtigste Regel: Nur wenn es wirklich kurz bleibt, ist `lambda` lesbar.


```python
quadrat_lambda = lambda x: x * x
print(quadrat_lambda(4))
```

Wichtig:
- In `lambda` steht nur **ein Ausdruck**.
- Fuer laengere Logik ist `def` lesbarer.


### 1.3 `lambda` mit `sorted`

Mit `key=` sagen wir `sorted`, **nach welchem Kriterium** sortiert werden soll.


```python
namen = ["Lea", "Alexander", "Bo", "Mira"]

# Sortierung nach Laenge des Namens
sortiert = sorted(namen, key=lambda name: len(name))
print(sortiert)
```

### 1.4 `lambda` mit `filter`

`filter(funktion, daten)` behaelt nur jene Elemente, bei denen die Funktion `True` liefert.
Das ist praktisch, wenn wir aus einer kleinen Liste einige Werte aussortieren wollen.


```python
zahlen = [7, -2, 0, 15, -9]

nur_positiv = list(filter(lambda x: x > 0, zahlen))
print(nur_positiv)
```

## 2. Rekursion

### 2.1 Warum benutzen wir Rekursion?

Rekursion ist sinnvoll, wenn ein Problem aus einer kleineren Version von sich selbst besteht.
Wir loesen dann nicht alles auf einmal, sondern Schritt fuer Schritt.

### 2.2 Die drei wichtigen Ideen

Bei Rekursion gibt es immer drei Teile:

1. **kleineres Problem**: Wir machen die Aufgabe kleiner.
2. **Basisfall**: Wir stoppen an einer ganz einfachen Stelle.
3. **rekursiver Schritt**: Wir rufen die Funktion mit der kleineren Aufgabe erneut auf.


```python
def countdown(n):
    if n == 0:              # Basisfall
        print("Start!")
        return

    print(n)
    countdown(n - 1)        # rekursiver Schritt

countdown(5)
```

### 2.3 Typischer Fehler: kein sauberer Basisfall

Fehlt der Basisfall, endet der Selbstaufruf nie. Dann kommt ein `RecursionError`.
Darum pruefen wir bei jeder rekursiven Funktion zuerst: Wann stoppt sie?


```python
def fehlerhaft(n):
    # WARNUNG: Diese Funktion stoppt nie sinnvoll.
    return fehlerhaft(n + 1)

# Nicht ausfuehren: fehlerhaft(0)
print("Diese Zelle zeigt nur das Problembeispiel.")
```

### 2.4 Beispiel: Fakultaet

`5! = 5 * 4 * 3 * 2 * 1`
Hier sieht man gut: Das Problem wird immer kleiner, bis der Basisfall erreicht ist.


```python
def fakultaet(n):
    if n <= 1:
        return 1
    return n * fakultaet(n - 1)

print(fakultaet(5))
```

### 2.5 Beispiel: Summe einer Liste

Wir nehmen immer das erste Element und addieren die rekursive Summe des Rests.


```python
def summe_rekursiv(liste):
    if len(liste) == 0:
        return 0
    return liste[0] + summe_rekursiv(liste[1:])

print(summe_rekursiv([4, 7, 1, 3]))
```

<span class="exercise_label">**Mini-Check 3:**</span>
Schreibe eine rekursive Funktion `enthaelt_wert(liste, ziel)`, die `True` zurueckgibt,
falls `ziel` in der Liste vorkommt, sonst `False`.
Tipp: Zuerst den Basisfall festlegen.


```python
def enthaelt_wert(liste, ziel):
    if len(liste) == 0:
        return False
    if liste[0] == ziel:
        return True
    return enthaelt_wert(liste[1:], ziel)

print(enthaelt_wert([10, 20, 30], 20))
print(enthaelt_wert([10, 20, 30], 99))
```

## 3. Zusammenfassung

| Thema | Kernidee | Mini-Regel |
|---|---|---|
| `lambda` | kurze anonyme Funktion | gut fuer kleine Ausdruecke |
| `sorted(..., key=...)` | Sortierkriterium separat angeben | `key` liefert Vergleichswert |
| `filter(...)` | Elemente per Bedingung behalten | Bedingung muss `True/False` liefern |
| Rekursion | Problem durch Selbstaufruf loesen | immer kleineres Problem + Basisfall + Schritt |

## Anwendung im Projekt: Sortieren und Summieren

Die Rechnungspositionen aus SW9 werden mit `sorted(..., key=lambda ...)` nach Betrag sortiert. Rekursion wird an einer kleinen Summe demonstriert: Der Basisfall ist die leere Liste, der rekursive Schritt nimmt den ersten Betrag und ruft sich für den Rest erneut auf. Für echte Massendaten ist die iterative Variante vorzuziehen.


```python
positionen = [{"positionsnummer": 1, "betrag": 300.0}, {"positionsnummer": 2, "betrag": 40.0}]
sortiert = sorted(positionen, key=lambda p: p["betrag"], reverse=True)

def summe_rekursiv(positionen):
    if not positionen:                 # Basisfall
        return 0.0
    return positionen[0]["betrag"] + summe_rekursiv(positionen[1:])  # rekursiver Schritt

def summe_iterativ(positionen):
    gesamt = 0.0
    for position in positionen:
        gesamt += position["betrag"]
    return gesamt

assert summe_rekursiv([]) == 0.0
assert summe_rekursiv([{"betrag": 7.5}]) == 7.5
assert summe_rekursiv(positionen) == summe_iterativ(positionen)
print(sortiert, summe_rekursiv(positionen))
```
