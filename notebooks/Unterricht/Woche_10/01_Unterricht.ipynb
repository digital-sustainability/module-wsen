## 1. Lambda-Funktionen

### 1.1 Kurzauffrischung: normale Funktion

Eine normale Funktion schreiben wir mit `def`. Sie hat einen Namen und kann mehrfach aufgerufen werden.


```python
def quadrat(x):
    return x * x

print(quadrat(4))
```

### 1.2 Was ist eine `lambda`-Funktion?

Eine `lambda`-Funktion ist eine kurze, anonyme Funktion.
Sie eignet sich fuer **kleine Berechnungen**, die wir nur direkt an einer Stelle brauchen.

Allgemeine Form:

`lambda parameter: ausdruck`


```python
quadrat_lambda = lambda x: x * x
print(quadrat_lambda(4))
```

Wichtig:
- In `lambda` steht nur **ein Ausdruck**.
- Fuer laengere Logik ist `def` lesbarer.


### 1.3 `lambda` mit `sorted`

Mit `key=` sagen wir `sorted`, **nach welchem Kriterium** sortiert werden soll.


```python
zahlen = [12, -5, 3, -20, 8]

# Sortierung nach Abstand zu 0
sortiert = sorted(zahlen, key=lambda x: abs(x))
print(sortiert)
```

### 1.4 `lambda` mit `filter`

`filter(funktion, daten)` behaelt nur jene Elemente, bei denen die Funktion `True` liefert.


```python
zahlen = [7, -2, 0, 15, -9]

nur_positiv = list(filter(lambda x: x > 0, zahlen))
print(nur_positiv)
```

## 2. Rekursion

### 2.1 Idee der Rekursion

Bei Rekursion ruft eine Funktion **sich selbst** auf.
Damit das endet, braucht es immer:

1. **Basisfall**: stoppt die Rekursion.
2. **Rekursiver Fall**: bringt das Problem einen Schritt Richtung Basisfall.


```python
def countdown(n):
    if n == 0:              # Basisfall
        print("Start!")
        return

    print(n)
    countdown(n - 1)        # Rekursiver Fall

countdown(5)
```

### 2.2 Typischer Fehler: kein sauberer Basisfall

Fehlt der Basisfall, endet der Selbstaufruf nie. Dann kommt ein `RecursionError`.


```python
def fehlerhaft(n):
    # WARNUNG: Diese Funktion stoppt nie sinnvoll.
    return fehlerhaft(n + 1)

# Nicht ausfuehren: fehlerhaft(0)
print("Diese Zelle zeigt nur das Problembeispiel.")
```

### 2.3 Beispiel: Fakultaet

`5! = 5 * 4 * 3 * 2 * 1`


```python
def fakultaet(n):
    if n <= 1:
        return 1
    return n * fakultaet(n - 1)

print(fakultaet(5))
```

### 2.4 Beispiel: Summe einer Liste

Wir nehmen immer das erste Element und addieren die rekursive Summe des Rests.


```python
def summe_rekursiv(liste):
    if len(liste) == 0:
        return 0
    return liste[0] + summe_rekursiv(liste[1:])

print(summe_rekursiv([4, 7, 1, 3]))
```

<span class="exercise_label">**Mini-Check 3:**</span>
Schreibe eine rekursive Funktion `enthaelt_wert(liste, ziel)`, die `True` zurueckgibt,
falls `ziel` in der Liste vorkommt, sonst `False`.


```python
def enthaelt_wert(liste, ziel):
    if len(liste) == 0:
        return False
    if liste[0] == ziel:
        return True
    return enthaelt_wert(liste[1:], ziel)

print(enthaelt_wert([10, 20, 30], 20))
print(enthaelt_wert([10, 20, 30], 99))
```

## 3. Zusammenfassung

| Thema | Kernidee | Mini-Regel |
|---|---|---|
| `lambda` | kurze anonyme Funktion | gut fuer kleine Ausdruecke |
| `sorted(..., key=...)` | Sortierkriterium separat angeben | `key` liefert Vergleichswert |
| `filter(...)` | Elemente per Bedingung behalten | Bedingung muss `True/False` liefern |
| Rekursion | Funktion ruft sich selbst | immer Basisfall + Schritt zum Basisfall |
