## 1. Lambda-Funktionen

### 1.1 Warum benutzen wir `lambda`?

Eine normale Funktion schreiben wir mit `def`. Das ist gut für längere Logik.
Eine `lambda` ist praktisch, wenn wir nur einen kurzen Ausdruck direkt übergeben wollen.


```python
def quadrat(x):
    return x * x

print(quadrat(4))
```

### 1.2 Was ist eine `lambda`-Funktion?

Eine `lambda`-Funktion ist eine kurze, anonyme Funktion.
Sie eignet sich für **kleine Berechnungen**, die wir nur direkt an einer Stelle brauchen.

Allgemeine Form:

`lambda parameter: ausdruck`

Die wichtigste Regel: Nur wenn es wirklich kurz bleibt, ist `lambda` lesbar.


```python
quadrat_lambda = lambda x: x * x
print(quadrat_lambda(4))
```

Wichtig:
- In `lambda` steht nur **ein Ausdruck**.
- Für längere Logik ist `def` lesbarer.


### 1.3 `lambda` mit `sorted`

Mit `key=` sagen wir `sorted`, **nach welchem Kriterium** sortiert werden soll.


```python
namen = ["Lea", "Alexander", "Bo", "Mira"]

# Sortierung nach Laenge des Namens
sortiert = sorted(namen, key=lambda name: name.lower())
print(sortiert)
```

### 1.4 `lambda` mit `filter`

`filter(funktion, daten)` behält nur jene Elemente, bei denen die Funktion `True` liefert.
Das ist praktisch, wenn wir aus einer kleinen Liste einige Werte aussortieren wollen.


```python
zahlen = [7, -2, 0, 15, -9]

nur_positiv = list(filter(lambda x: x > 0, zahlen))
print(nur_positiv)
```

## List-Slicing: Teile einer Liste auswählen

Mit **Slicing** wählen wir mehrere Elemente aus einer Liste aus. Das Ergebnis ist eine neue Liste; die ursprüngliche Liste bleibt unverändert.

Die allgemeine Schreibweise lautet:

`liste[start:stop:step]`

- `start`: Index des ersten Elements (wird eingeschlossen).
- `stop`: Index, **vor** dem das Slicing endet (wird nicht eingeschlossen).
- `step`: Abstand zwischen den ausgewählten Elementen.

`start`, `stop` und `step` können weggelassen werden. Standardmässig beginnt Python am Anfang, endet am Ende und verwendet den Schritt `1`.


```python
farben = ["rot", "grün", "blau", "gelb", "violett"]

print(farben[1:4])  # Index 1 bis vor Index 4
print(farben[:3])   # vom Anfang bis vor Index 3
print(farben[2:])   # ab Index 2 bis zum Ende
print(farben[:])    # eine Kopie der ganzen Liste
```

### Negative Indizes und Schrittweite

Negative Indizes zählen vom Ende der Liste: `-1` ist das letzte Element, `-2` das vorletzte. Mit `step` können wir Elemente überspringen. Ein negativer Schritt läuft rückwärts.


```python
zahlen = [10, 20, 30, 40, 50, 60]

print(zahlen[-3:])   # die letzten drei Elemente
print(zahlen[::2])   # jedes zweite Element
print(zahlen[::-1])  # eine umgekehrte Kopie
```

**Merke:** Der `stop`-Index ist immer ausgeschlossen. Ein Slice ausserhalb der gültigen Grenzen führt nicht zu einem `IndexError`; Python verwendet einfach die Elemente, die innerhalb der Liste liegen.

## 2. Rekursion

### 2.1 Warum benutzen wir Rekursion?

Rekursion ist sinnvoll, wenn ein Problem aus einer kleineren Version von sich selbst besteht.
Wir lösen dann nicht alles auf einmal, sondern Schritt für Schritt.

### 2.2 Die drei wichtigen Ideen

Bei Rekursion gibt es immer drei Teile:

1. **kleineres Problem**: Wir machen die Aufgabe kleiner.
2. **Basisfall**: Wir stoppen an einer ganz einfachen Stelle.
3. **rekursiver Schritt**: Wir rufen die Funktion mit der kleineren Aufgabe erneut auf.


```python
def countdown(n):
    if n == 0:              # Basisfall
        print("Start!")
        return

    print(n)
    countdown(n - 1)        # rekursiver Schritt

countdown(5)
```

### 2.3 Typischer Fehler: kein sauberer Basisfall

Fehlt der Basisfall, endet der Selbstaufruf nie. Dann kommt ein `RecursionError`.
Darum prüfen wir bei jeder rekursiven Funktion zuerst: Wann stoppt sie?


```python
def fehlerhaft(n):
    # WARNUNG: Diese Funktion stoppt nie sinnvoll.
    return fehlerhaft(n + 1)

# Nicht ausfuehren: fehlerhaft(0)
print("Diese Zelle zeigt nur das Problembeispiel.")
```

### 2.4 Beispiel: Fakultät

`5! = 5 * 4 * 3 * 2 * 1`
Hier sieht man gut: Das Problem wird immer kleiner, bis der Basisfall erreicht ist.


```python
def fakultaet(n):
    if n <= 1:
        return 1
    return n * fakultaet(n - 1)

print(fakultaet(5))
```

### 2.5 Beispiel: Summe einer Liste

Wir nehmen immer das erste Element und addieren die rekursive Summe des Rests.


```python
def summe_rekursiv(liste):
    if len(liste) == 0:
        return 0
    return liste[0] + summe_rekursiv(liste[1:])

print(summe_rekursiv([4, 7, 1, 3]))
```

## Anwendung im Projekt: Rechnungssumme rekursiv berechnen

Dieselben Positionen lassen sich auch rekursiv statt iterativ aufsummieren – als kleines Zusatzbeispiel zur eben eingeführten Rekursion.

```python
def summe_rekursiv(positionen):
    if not positionen:                 # Basisfall
        return 0.0
    return positionen[0]["betrag"] + summe_rekursiv(positionen[1:])  # rekursiver Schritt

def summe_iterativ(positionen):
    gesamt = 0.0
    for position in positionen:
        gesamt += position["betrag"]
    return gesamt

assert summe_rekursiv([]) == 0.0
assert summe_rekursiv([{"betrag": 7.5}]) == 7.5
assert summe_rekursiv(positionen) == summe_iterativ(positionen)
print(summe_rekursiv(positionen))
```

<span class="exercise_label">**Mini-Check 3:**</span>
Schreibe eine rekursive Funktion `enthaelt_wert(liste, ziel)`, die `True` zurückgibt,
falls `ziel` in der Liste vorkommt, sonst `False`.
Tipp: Zuerst den Basisfall festlegen.


```python
def enthaelt_wert(liste, ziel):
    if len(liste) == 0:
        return False
    if liste[0] == ziel:
        return True
    return enthaelt_wert(liste[1:], ziel)

print(enthaelt_wert([10, 20, 30], 20))
print(enthaelt_wert([10, 20, 30], 99))
```

## Zusammenfassung

| Thema | Kernidee | Mini-Regel |
|---|---|---|
| `lambda` | kurze anonyme Funktion | gut für kleine Ausdrücke |
| `sorted(..., key=...)` | Sortierkriterium separat angeben | `key` liefert Vergleichswert |
| `filter(...)` | Elemente per Bedingung behalten | Bedingung muss `True/False` liefern |
| List-Slicing | Teilbereiche als neue Liste auswählen | `stop` ist ausgeschlossen; `step` überspringt Elemente |
| Rekursion | Problem durch Selbstaufruf lösen | immer kleineres Problem + Basisfall + Schritt |
