## Aufgabe 1 (gefuehrt): Tuple-Grundlagen


```python
person = ("Mira", 19, "Informatik")

name = person[0]
alter = person[1]
fach = person[2]

punkt = (4, 9)
x, y = punkt  # Tuple-Unpacking

print(name, alter, fach)
print(x, y)
```

## Aufgabe 2 (gefuehrt): Dictionary-Grundlagen


```python
noten = {"Aylin": 5.0, "Noah": 4.5}

print("Ausgelesene Note:", noten["Aylin"])

noten["Noah"] = 5.0  # vorhandenen Eintrag aendern
noten["Lina"] = 5.5  # neuen Eintrag hinzufuegen

for schueler, note in noten.items():
    print(schueler, "->", note)
```

## Aufgabe 3 (teilweise gefuehrt): Dictionary umkehren


```python
def dictionary_umkehren(dictionary):
    neuer_dictionary = {}
    for key, value in dictionary.items():
        # Value wird zum neuen Key, Key wird zum neuen Value
        neuer_dictionary[value] = key
    return neuer_dictionary


print(dictionary_umkehren({"a": 1, "b": 2}))
```

## Aufgabe 4 (teilweise gefuehrt): Login-System


```python
def login(logins, eingabe_user, eingabe_passwort):
    # Erst pruefen, ob der User existiert
    if eingabe_user not in logins:
        return False

    # Dann Passwort vergleichen
    return logins[eingabe_user] == eingabe_passwort


print(login({"user1": "passwort1", "user2": "passwort2"}, "user1", "passwort1"))
print(login({"user1": "passwort1", "user2": "passwort2"}, "user3", "passwort1"))
print(login({"user1": "passwort1", "user2": "passwort2"}, "user1", "passwort3"))
```

## Aufgabe 5 (Transfer): Wort-Frequenzen


```python
def word_freq(text):
    haeufigkeiten = {}

    for wort in text.split():
        if wort in haeufigkeiten:
            haeufigkeiten[wort] += 1
        else:
            haeufigkeiten[wort] = 1

    return haeufigkeiten


print(word_freq("This is a simple example. This is a test example."))
```
