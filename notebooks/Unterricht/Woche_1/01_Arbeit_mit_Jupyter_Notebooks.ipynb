## Einführung

Jupyter Notebooks sind digitale Arbeitsblätter. Sie bestehen aus sogenannten „Zellen“. Eine Zelle kann entweder Text (Markdown) oder Python-Code enthalten.

Du führst eine Code-Zelle mit Shift+Enter oder mit dem Play-Button aus. Die Ausgabe erscheint direkt unter der Zelle.

Wenn du eine Zelle ausführst, arbeitet Python mit dem aktuellen Zustand des Notebooks. Das heißt: Werte, die du in einer früheren Zelle gespeichert hast, stehen in späteren Zellen wieder zur Verfügung. Wenn du die Zellen in einer anderen Reihenfolge ausführst, kann das Ergebnis anders aussehen.


### So arbeitest du mit einem Notebook

- Eine Zelle ist eine kleine Einheit im Notebook.
- Eine Code-Zelle enthält Python-Befehle.
- Eine Markdown-Zelle enthält Text, Erklärungen oder Notizen.
- Die Reihenfolge der Ausführung ist wichtig, weil der Notebook-Zustand gespeichert wird.

### Kurzglossar

- **Zelle**: eine einzelne Einheit im Notebook
- **Code**: Anweisungen, die Python ausführt
- **Ausgabe**: das Ergebnis einer Code-Zelle
- **Variable**: ein Name für einen Wert
- **Fehler**: eine Nachricht, dass Python etwas nicht verstehen oder ausführen konnte


## Markdown kurz erklärt
In Markdownzellen könnt ihr beliebigen Text einfügen, um zum Beispiel eure Notizen festzuhalten oder Code zu erklären. Die Markdownzelle könnt ihr mit einem Doppelklick bearbeiten. 

Markdown ist eine einfache Sprache, um Textdokumente zu formatieren. Überschriften beginnt man mit einem oder mehreren #, je mehr, desto kleiner wird die Überschrift:

### Kleine Überschrift
#### noch kleinere Überschrift

Auflistungen können einfach mit einem - erstellt werden:
- erster Punkt
- zweiter Punkt

oder mit Zahlen:
1. erster Punkt
2. zweiter Punkt

Um einen Umbruch zu erzwingen, müssen zwei "Returns" nacheinander gemacht werden: 
Dies ist kein Umbruch,

dies schon.

**Fetten Text** umgibt man mit zwei Sternchen, *kursiven Text* mit einem Sternchen. Will man Code schreiben, umgibt man ihn mit dem Symbol \` (accent grave), sodass er mit fester Breite formatiert wird: `dies sieht aus wie Code, wird aber nicht ausgeführt.`

Sobald man mit der Bearbeitung der Zelle fertig ist, kann man Shift + Enter drücken (oder auf den "Play"-Knopf klicken). Dann wird die Zelle formatiert.

## Code-Zellen
In Code-Zellen schreiben wir Python-Anweisungen. Wenn wir eine Code-Zelle ausführen, liest Python die Anweisung Zeile für Zeile und zeigt das Ergebnis direkt darunter an.

Ein erster Test:


````python
1 + 2
````

Nur das Resultat der letzten Code-Zeile wird ausgegeben:

````python
1 - 2
2 + 3
````

Der Zähler in eckigen Klammern links von der Zelle zeigt an, wann die Zelle zuletzt ausgeführt wurde. Wenn du dieselbe Zelle mehrfach ausführst, wird der Zähler größer.

Wichtig: Wenn du in einer früheren Zelle eine Variable definiert hast, bleibt ihr Wert im Notebook gespeichert. Das nennt man den Zustand des Kernels. Daher ist die Reihenfolge der Ausführung wichtig.


````python
a = 5
````

Der Wert 5 wurde in a gespeichert. Wenn wir nun schauen, was das Resultat von `a` ist, erhalten wir folgendes:

````python
a
````

Man muss aufpassen, wenn Variablen überschrieben werden: Wenn wir die Zellen in unterschiedlicher Reihenfolge ausführen, kann das Resultat anders sein. Hier ein kleines Beispiel: Führe zuerst die nächste Zelle aus, dann die übernächste, und dann nochmals die nächste Zelle. Das Resultat in der nächsten Zelle verändert sich.

````python
a + 2
````

````python
a = 7
````

Man sollte daher die Zellen immer von oben nach unten ausführen, wenn man ein Notebook nachvollziehen oder nacharbeiten möchte. So ist der Zustand leichter verständlich und die Ergebnisse bleiben reproduzierbar.
