Betrachte folgenden `while`-Loop:

```python
i = 5
while i < 15:
    i = i + 2
    if i != 11:
        print(i)

```

- a) Schreibe den `while`-Loop als `for`-Loop um
- b) Schreibe die Logik so um, dass `continue` verwendet wird.


````python
for i in range(5,16,2):
    if i == 11:
        continue
    print(i)
````

# 2.

Schreibe ein Programm, welches nach einer Zahl `n` fragt, und dann den Wert von `n! = (n * (n-1) * (n-2) * ... * 2 * 1)` zurückgibt.
<br>
Beispiel:
`3! = 3 * 2 * 1 = 6` 


````python
number = int(input("Gib eine ganze Zahl ein"))

result = 1

while number != 0:
    result = result * number
    number = number - 1
    
print(result)
````

# 3. 
Schreibe ein Programm, welche den Benutzer nach einer Zahl fragt. Anschliessend zeigt das Programm alle Quadratzahlen an, welche kleiner als die eingegebene Zahl sind.

Beispiel:

Ausgabe:

`Gib eine Zahl ein:`

Eingabe: 

`10`

Ausgabe: 
```
1 x 1 = 1
2 x 2 = 4
3 x 3 = 9
```

````python
number = int(input("Gib eine ganze Zahl ein"))

for i in range(1, number + 1):
    result = i * i
    if result >= number:
        break
    print(result)
````

# 4.

Schreibe ein Programm, welches nach einer Zahl fragt, und anschliessend die Quersumme dieser Zahl zurückgibt. Die Quersumme besteht aus der Summe aller Ziffern einer Zahl. Beispiel:

- Die Quersumme von `83` ist `8 + 3 = 11`
- Die Quersumme von `2519` ist `2 + 5 + 1 + 9 = 17`

````python
number = input("Gib eine ganze Zahl ein")

result = 0

for i in number:
    result = int(i) + result

print(result)
````

# 6.

Zahlen können sehr sensible Daten enthalten. 

Schreibe deshalb ein Programm, welches aus einem Text automatisch alle Zahlen rauslöscht. Dies machen wir, indem wir jede Ziffer mit einem X ersetzen. 

Beispiel:

```
Wir verfügen über ein Vermögen von CHF 8'230'521.23
Wir verfügen über ein Vermögen von CHF X'XXX'XXX.XX
```

Oder:
```
Wir verkaufen die Aktie am 23.09.
Wir verkaufen die Aktie am XX.XX.
```


````python
text = input("Gib einen Text ein")

result = ""

for i in text:
    if i in "0123456789":
        result = result + "X"
    else:
        result = result + i

print(result)
````

````python
# Aussage: Es gibt eine blaue Kugel
result = False
for x in range(11):
    result = (x in Blue) or result

print (result)
````

````python
# Aussage: Wenn eine Kugel weiss ist, muss sie eine orange Kugel berühren
result1 = True
for x in range (11):
    if x in White:
        result2 = False
        for y in range(11):
            result2 = result2 or (y in Orange and (x,y) in T)
        result1 = result1 and result2

print(result1)
````

````python
# Aussage: Schwarze Kugeln berühren orange Kugeln nicht
result = True
for x in range(11):
    for y in range(11):
        result = result and not (x in Black and y in Orange and (x,y) in T)

print(result)
````

````python
# Aussage: Grüne Kugeln nicht unter anderen Kugeln
result2 = True
for x in range(11):
    if x in Green:
        result1 = True
        for y in range(11):
            result1 = result1 and ((x,y) not in B)
        result2 = result2 and result1

print(result2)
````
