## Loesung 1: `Produkt`
Typischer Fehler: Prozent als Ganzzahl ohne Division durch 100 verwenden.

```python
class Produkt:
    def __init__(self, produkt_nr, name, preis):
        self.produkt_nr = produkt_nr
        self.name = name
        self.preis = preis

    def rabatt_preis(self, prozent):
        return self.preis * (1 - prozent / 100)

produkt = Produkt(10, "Maus", 50)
produkt.rabatt_preis(20)
```

## Loesung 2: `Warenkorb`
Typischer Fehler: `self.produkte` vergessen oder nicht als Liste initialisieren.

```python
class Warenkorb:
    def __init__(self):
        self.produkte = []

    def produkt_hinzufuegen(self, produkt):
        self.produkte.append(produkt)

    def gesamtpreis(self):
        total = 0
        for produkt in self.produkte:
            total += produkt.preis
        return total

p1 = Produkt(1, "Tastatur", 30)
p2 = Produkt(2, "Headset", 45)

korb = Warenkorb()
korb.produkt_hinzufuegen(p1)
korb.produkt_hinzufuegen(p2)
korb.gesamtpreis()
```

## Loesung 3: BO <-> ERD (1:n)
Typischer Fehler: Eine einzelne Bestellung statt einer Liste speichern.

```python
class Bestellung:
    def __init__(self, bestell_nr, betrag):
        self.bestell_nr = bestell_nr
        self.betrag = betrag

class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name
        self.bestellungen = []

    def bestellung_hinzufuegen(self, bestellung):
        self.bestellungen.append(bestellung)

kunde = Kunde(7, "Lina")
kunde.bestellung_hinzufuegen(Bestellung(9001, 120))
kunde.bestellung_hinzufuegen(Bestellung(9002, 35))

len(kunde.bestellungen), [b.bestell_nr for b in kunde.bestellungen]
```

## Kurz-Reflexion
- `Kunde` und `Bestellung` sind BOs im Code.
- Im ERD sind sie Entitaeten.
- Die Liste `bestellungen` zeigt die 1:n Beziehung im Programm.
