## 1. Warum OOP? Ein Alltagsvergleich

Stell dir vor, du entwickelst ein System für einen Online-Shop.  
Für jeden Kunden müsstest du ohne OOP einzelne Variablen anlegen:

```python
kunde_1_name  = "Mia Keller"
kunde_1_email = "mia@example.ch"
kunde_2_name  = "Noah Berger"
kunde_2_email = "noah@example.ch"
# ... und so weiter für 1000 Kunden
```

Das wird schnell unübersichtlich. Mit OOP fasst du alles, was zu *einem Kunden* gehört, in einer einzigen **Klasse** zusammen – wie ein Formular, das du für jeden neuen Kunden ausfüllst.


---
## 2. Die vier Grundbegriffe

| Begriff | Was ist es? | Alltagsvergleich | Python-Beispiel |
|---|---|---|---|
| **Klasse** | Der Bauplan / das Formular | Kfz-Fahrzeugschein-Vorlage | `class Kunde:` |
| **Objekt** | Ein konkretes Exemplar der Klasse | Der ausgefüllte Fahrzeugschein von *einem* Auto | `kunde_1 = Kunde(...)` |
| **Attribut** | Eine gespeicherte Eigenschaft des Objekts | Feld "Fahrername" auf dem Schein | `self.name = "Mia"` |
| **Methode** | Eine Funktion, die zum Objekt gehört | "Ummelden" als Aktion auf dem Schein | `def email_aendern(self, ...)` |

> 💡 **Merkregel:** Die Klasse ist der Bauplan, das Objekt ist das fertige Produkt.

## 3. Eine erste, leere Klasse

Wir beginnen mit dem einfachstmöglichen Fall: einer Klasse ohne Inhalt.  
Das Schlüsselwort `class` gefolgt vom Namen und einem Doppelpunkt eröffnet die Klassendefinition.  
`pass` bedeutet: *«Diese Klasse ist absichtlich noch leer.»*


````python
class Kunde:
    pass  # leere Klasse – noch kein Inhalt

# Ein Objekt dieser Klasse erzeugen:
kunde_1 = Kunde()

# Welchen Typ hat das Objekt?
type(kunde_1)
````

**Was zeigt `type(kunde_1)`?**  
`<class '__main__.Kunde'>` – Python bestätigt, dass `kunde_1` ein Objekt der Klasse `Kunde` ist.

> **Namenskonvention:** Klassennamen beginnen immer mit einem **Grossbuchstaben** (`Kunde`, `Bestellung`).  
> Variablen und Methoden bleiben kleingeschrieben (`kunden_nr`, `email_aendern`).

## 4. `__init__` – der Constructor

Bisher kann unsere Klasse nichts speichern.  
Mit der speziellen Methode `__init__` (sprich: *dunder init*) definieren wir, **welche Daten ein neues Objekt beim Erstellen mitbekommt**.

`__init__` wird automatisch von Python aufgerufen, sobald wir `Kunde(...)` schreiben.  
Deshalb heisst sie auch **Constructor** – sie *baut* das Objekt auf.

### Aufbau Schritt für Schritt

```
class Kunde:
    def __init__(self, kunden_nr, name, email):
        #          ^^^   ^^^^^^^^^ ^^^^  ^^^^^
        #          (1)      (2)    (3)   (4)
```

| # | Was steht da? | Bedeutung |
|---|---|---|
| (1) | `self` | Das Objekt selbst – dazu gleich mehr |
| (2) | `kunden_nr` | Erster Eingabewert beim Erstellen |
| (3) | `name` | Zweiter Eingabewert |
| (4) | `email` | Dritter Eingabewert |


````python
class Kunde:
    def __init__(self, kunden_nr, name, email):
        # Mit self.xxx = ... speichern wir die Werte ALS ATTRIBUTE am Objekt
        self.kunden_nr = kunden_nr
        self.name      = name
        self.email     = email

# Ein neues Kunden-Objekt anlegen:
kunde_a = Kunde(1001, "Mia Keller", "mia@example.ch")

# Auf die Attribute zugreifen:
print("Name: ",  kunde_a.name)
print("E-Mail:", kunde_a.email)
````

## 5. Was ist `self`?

`self` ist der Weg, wie eine Methode *weiss, zu welchem Objekt sie gerade gehört*.

**Analogie:** Stell dir vor, du bist in einem Büro mit zehn Mitarbeitenden.  
Jemand ruft «Bitte schreibe deinen Namen auf das Formular!»  
Jede Person versteht «deinen» als sich selbst – nicht als die Person daneben.  
Genau das macht `self` in Python.

**Technisch:**  
- `self` ist immer der *erste Parameter* einer Methode.
- Du musst ihn beim Aufrufen **nicht** mitgeben – Python übergibt `self` automatisch.
- Innerhalb der Methode greifst du über `self.attribut` auf die gespeicherten Daten des Objekts zu.


````python
# Demonstriere self: Zwei Objekte – gleiche Klasse, eigene Daten
kunde_b = Kunde(1002, "Noah Berger", "noah@example.ch")
kunde_c = Kunde(1003, "Sara Meier",  "sara@example.ch")

# self zeigt auf das jeweilige Objekt – jedes hat seinen eigenen Namen
print(kunde_b.name)  # Noah Berger
print(kunde_c.name)  # Sara Meier
````

> 💡 `kunde_b` und `kunde_c` sind **zwei verschiedene Objekte** derselben Klasse.  
> Jedes hat seine eigenen Attributwerte – sie teilen nur den Bauplan.

## 6. Methoden: Verhalten ins Objekt einbauen

Bisher speichert `Kunde` nur Daten.  
Methoden erlauben es, **Aktionen** zu definieren, die das Objekt ausführen kann.

Wir fügen der Klasse `Konto` zwei Methoden hinzu:
- `einzahlen(betrag)` – erhöht den Saldo
- `abheben(betrag)` – verringert den Saldo

Beachte: Methoden haben `self` als ersten Parameter – genau wie `__init__`.


````python
class Konto:
    def __init__(self, inhaber, saldo=0):
        # saldo=0 ist ein Standardwert: wird kein Saldo angegeben, startet das Konto mit 0
        self.inhaber = inhaber
        self.saldo   = saldo

    def einzahlen(self, betrag):
        # self.saldo += betrag   bedeutet:  self.saldo = self.saldo + betrag
        self.saldo += betrag

    def abheben(self, betrag):
        self.saldo -= betrag
````

````python
# Schritt für Schritt nachvollziehen:
konto = Konto("Mia Keller", saldo=50)
print("Start-Saldo:     ", konto.saldo)  # 50

konto.einzahlen(30)
print("Nach Einzahlung: ", konto.saldo)  # 80

konto.abheben(20)
print("Nach Abhebung:   ", konto.saldo)  # 60
````

**Was passiert bei `konto.einzahlen(30)`?**

1. Python sucht die Methode `einzahlen` in der Klasse `Konto`.
2. Es übergibt automatisch das Objekt `konto` als `self`.
3. `betrag` bekommt den Wert `30`.
4. `self.saldo += betrag` verändert den Saldo *dieses* Objekts.

> 💡 Die Methode wird immer mit einem Punkt aufgerufen: `konto.einzahlen(30)` – **nicht** `einzahlen(konto, 30)`.

## 7. Vom Code zum Businessobjekt (BO) und zum ERD

### Was ist ein Businessobjekt?

Ein **Businessobjekt (BO)** ist ein Konzept aus der realen Geschäftswelt (z.B. «Kunde», «Bestellung», «Produkt»), das wir im Code als Klasse abbilden.

### Was ist ein ERD?

Ein **Entity-Relationship-Diagramm (ERD)** zeigt, welche BOs existieren, welche Eigenschaften sie haben und wie sie zusammenhängen – **ohne Code**, nur als Diagramm.

### Die Brücke: Code ↔ BO ↔ ERD

| Code-Element | BO-Begriff | ERD-Element |
|---|---|---|
| `class Kunde` | Businessobjekt «Kunde» | Entität `Kunde` |
| `self.kunden_nr` | Eindeutiger Bezeichner | Primärschlüssel |
| `self.name`, `self.email` | Eigenschaften | Attribute der Entität |
| `class Bestellung` | Businessobjekt «Bestellung» | Entität `Bestellung` |
| Liste `self.bestellungen` in `Kunde` | Ein Kunde hat *mehrere* Bestellungen | Beziehung **1 : n** |

### ERD-Skizze (als Text)

```
Kunde                     Bestellung
─────────────────         ─────────────────
kunden_nr  (PK)  1 ──── n  bestell_nr  (PK)
name                       betrag
email
```

Das `1 ──── n` bedeutet: **Ein** Kunde kann **mehrere** Bestellungen haben.


````python
# Im Code sieht diese 1:n-Beziehung so aus:

class Bestellung:
    def __init__(self, bestell_nr, betrag):
        self.bestell_nr = bestell_nr
        self.betrag     = betrag


class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr   = kunden_nr
        self.name        = name
        # Eine leere Liste nimmt die Bestellungen auf (= die «n»-Seite der Beziehung)
        self.bestellungen = []

    def bestellung_hinzufuegen(self, bestellung):
        # Bestellung zur Liste des Kunden hinzufügen
        self.bestellungen.append(bestellung)


# Nutzung:
kunde = Kunde(1, "Noah Berger")
kunde.bestellung_hinzufuegen(Bestellung(501, 79.90))
kunde.bestellung_hinzufuegen(Bestellung(502, 14.50))

print("Anzahl Bestellungen:", len(kunde.bestellungen))
print("Erste Bestellung, Betrag:", kunde.bestellungen[0].betrag)
````

**Was macht `kunde.bestellungen[0].betrag`?**

1. `kunde.bestellungen` → die Liste aller Bestellungen des Kunden
2. `[0]` → das erste Element der Liste (eine `Bestellung`-Objekt)
3. `.betrag` → das Attribut `betrag` dieses Bestellungs-Objekts

Das nennt man **Verkettung** – man greift von einem Objekt über ein Attribut auf ein anderes Objekt zu.

| Was haben wir gelernt? | Kurz erklärt |
|---|---|
| `class Name:` | Definiert den Bauplan einer Klasse |
| `__init__(self, ...)` | Wird beim Erstellen eines Objekts automatisch aufgerufen; legt Attribute fest |
| `self` | Verweist auf das aktuelle Objekt; immer erster Parameter einer Methode |
| `self.x = ...` | Speichert einen Wert als Attribut am Objekt |
| Methode `def m(self, ...)` | Funktion, die zum Objekt gehört; wird mit `objekt.m(...)` aufgerufen |
| 1:n-Beziehung | Im Code: eine Liste im Attribut; im ERD: Verbindung mit `1 ──── n` |
