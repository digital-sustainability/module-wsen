## 1. Eltern- und Kindklasse erkennen

### 1.1 Fahrzeug und Auto

```python
class Fahrzeug:
    def __init__(self, marke, tempo):
        self.marke = marke
        self.tempo = tempo

    def beschreibung(self):
        return f"{self.marke} faehrt {self.tempo} km/h"


class Auto(Fahrzeug):
    def hupe(self):
        return "beep"


a = Auto("Volvo", 120)
print(a.beschreibung())
print(a.hupe())
```

### 1.2 Velo als einfache Erweiterung
`Velo` erbt `marke`, `tempo` und `beschreibung()` von `Fahrzeug`.
Das neue Attribut `gaenge` wird nur in der Kindklasse gesetzt.

```python
class Velo(Fahrzeug):
    def __init__(self, marke, tempo, gaenge):
        super().__init__(marke, tempo)
        self.gaenge = gaenge


v = Velo("BMC", 30, 12)
print(v.beschreibung())
print(v.gaenge)
```

## 2. `super()` in Konstruktoren und Methoden

### 2.1 Person und Student

```python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter


class Student(Person):
    def __init__(self, name, alter, studiengang):
        super().__init__(name, alter)
        self.studiengang = studiengang


s = Student("Lea", 22, "Wirtschaftsinformatik")
print(s.name, s.alter, s.studiengang)
```

### 2.2 Vorstellen erweitern
Die Elternmethode liefert den Grundsatz, die Kindklasse fuegt nur den Studiengang dazu.
So bleibt die gemeinsame Logik an einer Stelle.

```python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    def vorstellen(self):
        return f"Ich bin {self.name} und {self.alter} Jahre alt."


class Student(Person):
    def __init__(self, name, alter, studiengang):
        super().__init__(name, alter)
        self.studiengang = studiengang

    def vorstellen(self):
        basis = super().vorstellen()
        return f"{basis} Ich studiere {self.studiengang}."


s = Student("Lea", 22, "Wirtschaftsinformatik")
print(s.vorstellen())
```

## 3. Ergaenzen oder ueberschreiben

### 3.1 Bus als Spezialisierung

```python
class Bus(Fahrzeug):
    def __init__(self, marke, tempo, plaetze):
        super().__init__(marke, tempo)
        self.plaetze = plaetze

    def beschreibung(self):
        return f"{super().beschreibung()} und hat {self.plaetze} Plaetze"


b = Bus("MAN", 90, 52)
print(b.beschreibung())
```

### 3.2 Schulpersonen-Hierarchie
Hier wird `vorstellen()` in der Kindklasse erweitert: zuerst kommt die Elternlogik, dann die neue Info zum Fach.

```python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    def vorstellen(self):
        return f"Ich bin {self.name} und {self.alter} Jahre alt."


class Lehrperson(Person):
    def __init__(self, name, alter, fach):
        super().__init__(name, alter)
        self.fach = fach

    def vorstellen(self):
        basis = super().vorstellen()
        return f"{basis} Ich unterrichte {self.fach}."


lp = Lehrperson("Herr Meier", 41, "Informatik")
print(lp.vorstellen())
```
