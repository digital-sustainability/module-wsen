## 1. Grundlagen der Vererbung

### 1.1 Fahrzeug und Velo

```python
class Fahrzeug:
    def __init__(self, marke, tempo):
        self.marke = marke
        self.tempo = tempo

    def beschreibung(self):
        return f"{self.marke} faehrt {self.tempo} km/h"


class Velo(Fahrzeug):
    def __init__(self, marke, tempo, gaenge):
        super().__init__(marke, tempo)
        self.gaenge = gaenge

    def klingel(self):
        return "ring"


v = Velo("BMC", 30, 12)
print(v.beschreibung())
print(v.klingel())
```

### 1.2 Bus als Spezialisierung

```python
class Bus(Fahrzeug):
    def __init__(self, marke, tempo, plaetze):
        super().__init__(marke, tempo)
        self.plaetze = plaetze

    def beschreibung(self):
        return f"{self.marke} faehrt {self.tempo} km/h und hat {self.plaetze} Plaetze"


b = Bus("MAN", 90, 52)
print(b.beschreibung())
```

## 2. `super()` und Konstruktoren

### 2.1 Person und Student

```python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter


class Student(Person):
    def __init__(self, name, alter, studiengang):
        super().__init__(name, alter)
        self.studiengang = studiengang


s = Student("Lea", 22, "Wirtschaftsinformatik")
print(s.name, s.alter, s.studiengang)
```

### 2.2 Methode erweitern mit `super()`

```python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    def vorstellen(self):
        return f"Ich bin {self.name} und {self.alter} Jahre alt."


class Student(Person):
    def __init__(self, name, alter, studiengang):
        super().__init__(name, alter)
        self.studiengang = studiengang

    def vorstellen(self):
        basis = super().vorstellen()
        return f"{basis} Ich studiere {self.studiengang}."


s = Student("Lea", 22, "Wirtschaftsinformatik")
print(s.vorstellen())
```

## 3. Polymorphie mit Vererbung

### 3.1 Tierhierarchie

```python
class Tier:
    def __init__(self, name):
        self.name = name

    def geraeusch(self):
        return "(kein spezifisches Geraeusch)"


class Hund(Tier):
    def geraeusch(self):
        return "wuff"


class Katze(Tier):
    def geraeusch(self):
        return "miau"


class Ente(Tier):
    def geraeusch(self):
        return "quak"


tiere = [Hund("Balu"), Katze("Luna"), Ente("Daisy")]
for t in tiere:
    print(t.name, t.geraeusch())
```

### 3.2 Einheitliche Verarbeitung

```python
def drucke_geraeusche(tiere):
    for tier in tiere:
        print(f"{tier.name}: {tier.geraeusch()}")


print("Ausgabe ueber Funktion:")
drucke_geraeusche(tiere)
```
