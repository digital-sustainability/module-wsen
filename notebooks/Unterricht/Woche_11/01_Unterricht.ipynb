## 1. Vererbung: Grundidee

Mit Vererbung uebernimmt eine Kindklasse Eigenschaften und Methoden einer Elternklasse.

Merksatz: Eine Kindklasse ist eine spezialisierte Form der Elternklasse.

```python
class Fahrzeug:
    def __init__(self, marke, tempo):
        self.marke = marke
        self.tempo = tempo

    def beschreibung(self):
        return f"{self.marke} faehrt {self.tempo} km/h"

f = Fahrzeug("Generic", 40)
print(f.beschreibung())
```

### 1.1 Kindklasse definieren

Eine Kindklasse steht in Klammern hinter dem Klassennamen.

```python
class Auto(Fahrzeug):
    def hupe(self):
        return "beep"

a = Auto("Volvo", 120)
print(a.beschreibung())  # von Fahrzeug geerbt
print(a.hupe())          # nur in Auto
```

## 2. Methoden ueberschreiben (Override)

Kindklassen duerfen geerbte Methoden anpassen.

```python
class SportAuto(Fahrzeug):
    def beschreibung(self):
        return f"Sportauto {self.marke}: {self.tempo} km/h"

s = SportAuto("Porsche", 280)
print(s.beschreibung())
```

### 2.1 `super()` nutzen

Mit `super()` kannst du Logik der Elternklasse wiederverwenden, statt alles neu zu schreiben.

```python
class ElektroAuto(Fahrzeug):
    def __init__(self, marke, tempo, akku_prozent):
        super().__init__(marke, tempo)
        self.akku_prozent = akku_prozent

    def beschreibung(self):
        basis = super().beschreibung()
        return f"{basis} (Akku: {self.akku_prozent}%)"

e = ElektroAuto("Tesla", 160, 78)
print(e.beschreibung())
```

## 3. Kleine Hierarchie

Vererbung ist besonders nuetzlich, wenn mehrere Klassen gemeinsames Verhalten teilen.

```python
class Tier:
    def __init__(self, name):
        self.name = name

    def geraeusch(self):
        return "?"

class Hund(Tier):
    def geraeusch(self):
        return "wuff"

class Katze(Tier):
    def geraeusch(self):
        return "miau"

tiere = [Hund("Balu"), Katze("Luna")]
for tier in tiere:
    print(f"{tier.name}: {tier.geraeusch()}")
```

### Live-Uebung A: Schulpersonen-Hierarchie

Ziel: Verstehen, wann eine Kindklasse nur erweitert und wann sie ueberschreibt.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def rolle(self):
        return "Person"

    def vorstellen(self):
        return f"Ich bin {self.name} ({self.rolle()})."

class Student(Person):
    def __init__(self, name, klasse):
        super().__init__(name)
        self.klasse = klasse

    def rolle(self):
        return "Student"

class Lehrperson(Person):
    def __init__(self, name, fach):
        super().__init__(name)
        self.fach = fach

    def rolle(self):
        return "Lehrperson"

    def vorstellen(self):
        basis = super().vorstellen()
        return f"{basis} Ich unterrichte {self.fach}."

# Mini-Auftraege im Plenum:
# 1) Erzeuge je ein Objekt von Student und Lehrperson.
# 2) Gib jeweils vorstellen() aus.
# 3) Erweitere Student um ein Attribut "schwerpunkt" und nimm es in vorstellen() auf.
```

### Live-Uebung B: Fehlersuche bei Vererbung

Ziel: Typische Fehler mit Konstruktoren und `super()` erkennen und korrigieren.

```python
class Konto:
    def __init__(self, inhaber, saldo):
        self.inhaber = inhaber
        self.saldo = saldo

    def info(self):
        return f"{self.inhaber}: CHF {self.saldo}"

class Sparkonto(Konto):
    def __init__(self, inhaber, saldo, zinssatz):
        # BUG: super().__init__(...) fehlt
        self.zinssatz = zinssatz

    def info(self):
        # BUG-Folge: self.inhaber / self.saldo sind ohne super() nicht gesetzt
        return f"{super().info()} | Zins: {self.zinssatz}%"

# Auftrag:
# 1) Fuehre den Code aus und lies die Fehlermeldung genau.
# 2) Behebe den Fehler im Konstruktor.
# 3) Teste mit: s = Sparkonto("Mila", 1500, 1.2); print(s.info())
```
