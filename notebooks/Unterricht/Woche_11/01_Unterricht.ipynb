## 1. Vererbung: Die Grundidee

Vererbung bedeutet: Eine Kindklasse bekommt Eigenschaften und Methoden von einer Elternklasse.
Das ist praktisch, wenn zwei Klassen viel gemeinsam haben.

Merksatz: Erst die gemeinsame Basis bauen, dann nur das Neue in der Kindklasse ergänzen.

```python
class Fahrzeug:
    def __init__(self, marke, tempo):
        self.marke = marke
        self.tempo = tempo

    def beschreibung(self):
        return f"{self.marke} faehrt {self.tempo} km/h"


f = Fahrzeug("Generic", 40)
print(f.beschreibung())
```

### 1.1 Kindklasse definieren

Eine Kindklasse steht in Klammern hinter dem Klassennamen.
Sie bekommt die geerbten Attribute und Methoden automatisch mit.

```python
class Auto(Fahrzeug):
    def hupe(self):
        return "beep"


a = Auto("Volvo", 120)
print(a.beschreibung())  # geerbt von Fahrzeug
print(a.hupe())          # neu in Auto
```


## Anwendung im Projekt: Rechnungstypen durch Vererbung

`Rechnung` definiert gemeinsame Kundendaten, Positionen und `gesamtbetrag()`. Eine `Gutschrift` erbt diese Struktur und überschreibt die Betragslogik: Sie weist den Betrag als negativen Wert aus. So bleiben gemeinsame Daten und Methoden wiederverwendbar, während der fachliche Unterschied klar sichtbar wird.


```python
class Rechnung:
    def __init__(self, nummer, kunde, positionen):
        self.nummer, self.kunde, self.positionen = nummer, kunde, positionen
    def gesamtbetrag(self):
        return sum(p["betrag"] for p in self.positionen)
    def ausgabe_typ(self):
        return "Rechnung"

class Gutschrift(Rechnung):
    def gesamtbetrag(self):
        return -super().gesamtbetrag()
    def ausgabe_typ(self):
        return "Gutschrift"

positionen = [{"betrag": 340.0}]
assert Rechnung("R-1", "K-100", positionen).gesamtbetrag() == 340.0
assert Gutschrift("G-1", "K-100", positionen).gesamtbetrag() == -340.0
print(Gutschrift("G-1", "K-100", positionen).ausgabe_typ())
```

## 2. Ergänzen oder überschreiben

Vererbung ist nicht nur Wiederverwendung. Manchmal soll die Kindklasse Verhalten **ergänzen**.
Manchmal soll sie Verhalten **überschreiben** und etwas anderes machen.

```python
class SportAuto(Fahrzeug):
    def beschreibung(self):
        return f"Sportauto {self.marke}: {self.tempo} km/h"


s = SportAuto("Porsche", 280)
print(s.beschreibung())
```

### 2.1 `super()` nutzen

`super()` bedeutet: Reuse den Code der Elternklasse, dann füge in der Kindklasse etwas Neues hinzu.
Das ist besonders hilfreich bei Konstruktoren und bei erweiterten Methoden.

```python
class ElektroAuto(Fahrzeug):
    def __init__(self, marke, tempo, akku_prozent):
        super().__init__(marke, tempo)
        self.akku_prozent = akku_prozent

    def beschreibung(self):
        basis = super().beschreibung()
        return f"{basis} (Akku: {self.akku_prozent}%, elektrisch)"


e = ElektroAuto("Tesla", 160, 78)
print(e.beschreibung())
```

## 3. Live-Anwendung: Schulpersonen-Hierarchie

Hier siehst du beide Ideen zusammen:
- `Student` erweitert `Person`
- `Lehrperson` erweitert `Person` und passt `vorstellen()` an

So wird sichtbar, wann eine Kindklasse nur ergänzt und wann sie das Verhalten überschreibt.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def vorstellen(self):
        return f"Ich bin {self.name}."


class Student(Person):
    def __init__(self, name, klasse):
        super().__init__(name)
        self.klasse = klasse

    def vorstellen(self):
        return f"{super().vorstellen()} Ich gehe in die Klasse {self.klasse}."


class Lehrperson(Person):
    def __init__(self, name, fach):
        super().__init__(name)
        self.fach = fach

    def vorstellen(self):
        return f"{super().vorstellen()} Ich unterrichte {self.fach}."


personen = [Student("Mila", "2A"), Lehrperson("Herr Meier", "Informatik")]
for person in personen:
    print(person.vorstellen())
```
