Tipp: Arbeite die Unteraufgaben von oben nach unten ab. Jede Funktion baut auf der vorherigen auf.

```python
import random

OPTIONEN = ["schere", "stein", "papier"]

def wahl_computer():
    """Der Computer wählt eine der drei Optionen zufällig."""
    return random.choice(OPTIONEN)
```
