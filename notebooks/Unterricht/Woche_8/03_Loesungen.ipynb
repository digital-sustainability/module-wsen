## Loesung 1: Begriffe erkennen

- **Klasse:** `Kunde`
- **Objekt:** `kunde`
- **Attribut-Zeilen:** `self.kunden_nr = kunden_nr` und `self.name = name`
- **Methode:** `def vorstellen(self):`

Die Methode gehoert zum Objekt, darum wird sie mit `kunde.vorstellen()` aufgerufen.


```python
class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name

    def vorstellen(self):
        return f"Hallo, ich bin {self.name}."

kunde = Kunde(1001, "Mia Keller")
kunde.vorstellen()
```

## Loesung 3: Businessobjekt und ERD (1:n)

Typischer Fehler: Nur eine Bestellung speichern statt einer Liste.


```python
class Bestellung:
    def __init__(self, bestell_nr, betrag):
        self.bestell_nr = bestell_nr
        self.betrag = betrag

class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name
        self.bestellungen = []

    def bestellung_hinzufuegen(self, bestellung):
        self.bestellungen.append(bestellung)

kunde = Kunde(7, "Lina")
kunde.bestellung_hinzufuegen(Bestellung(9001, 120))
kunde.bestellung_hinzufuegen(Bestellung(9002, 35))

len(kunde.bestellungen), [b.bestell_nr for b in kunde.bestellungen]
```

## Musterloesung: Rechnung als Businessobjekte
Die Klassen speichern ihre Daten in Attributen. Die Berechnung erfolgt ueber normale Methoden wie `summe()` und `netto()` statt ueber `@property`.


```python
class Kunde:
    def __init__(self, kunden_id, name):
        self.kunden_id = kunden_id
        self.name = name
        self.rechnungen = []

class Artikel:
    def __init__(self, artikel_id, bezeichnung, einzelpreis, mwst_satz):
        self.artikel_id = artikel_id
        self.bezeichnung = bezeichnung
        self.einzelpreis = einzelpreis
        self.mwst_satz = mwst_satz

class Rechnungsposition:
    def __init__(self, artikel, menge):
        self.artikel = artikel
        self.menge = menge

    def summe(self):
        return round(self.menge * self.artikel.einzelpreis, 2)

class Rechnung:
    def __init__(self, rechnung_id, kunde):
        self.rechnung_id = rechnung_id
        self.kunde = kunde
        self.positionen = []
        kunde.rechnungen.append(self)

    def position_hinzufuegen(self, position):
        self.positionen.append(position)

    def netto(self):
        total = 0
        for position in self.positionen:
            total += position.summe()
        return round(total, 2)

kunde = Kunde(1, "Alpen Technik AG")
kabel = Artikel(1001, "USB-C Kabel", 12.50, 7.7)
buch = Artikel(1002, "Notizbuch Recycling", 8.90, 2.5)
rechnung = Rechnung(1, kunde)
rechnung.position_hinzufuegen(Rechnungsposition(kabel, 2))
rechnung.position_hinzufuegen(Rechnungsposition(buch, 1))

print(kunde.name, rechnung.netto(), "CHF", len(rechnung.positionen), "Positionen")
assert rechnung.netto() == 33.90
```

## Kurz-Reflexion
- `Kunde` und `Bestellung` sind Businessobjekte im Code.
- Im ERD sind sie Entitaeten.
- Die Liste `bestellungen` zeigt die 1:n-Beziehung.
