## Loesung 1: Begriffe erkennen

- **Klasse:** `Kunde`
- **Objekt:** `kunde`
- **Attribut-Zeilen:** `self.kunden_nr = kunden_nr` und `self.name = name`
- **Methode:** `def vorstellen(self):`

Die Methode gehoert zum Objekt, darum wird sie mit `kunde.vorstellen()` aufgerufen.


```python
class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name

    def vorstellen(self):
        return f"Hallo, ich bin {self.name}."

kunde = Kunde(1001, "Mia Keller")
kunde.vorstellen()
```

## Loesung 2: `Produkt`

Typischer Fehler: Attribute nicht im Objekt speichern oder vergessen, einen Wert mit `return` zurueckzugeben.


```python
class Produkt:
    def __init__(self, produkt_nr, name, preis):
        self.produkt_nr = produkt_nr
        self.name = name
        self.preis = preis

    def kurzinfo(self):
        return f"{self.name} kostet {self.preis} CHF."

produkt = Produkt(10, "Maus", 50)
produkt.kurzinfo()
```

## Loesung 3: Businessobjekt und ERD (1:n)

Typischer Fehler: Nur eine Bestellung speichern statt einer Liste.


```python
class Bestellung:
    def __init__(self, bestell_nr, betrag):
        self.bestell_nr = bestell_nr
        self.betrag = betrag

class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name
        self.bestellungen = []

    def bestellung_hinzufuegen(self, bestellung):
        self.bestellungen.append(bestellung)

kunde = Kunde(7, "Lina")
kunde.bestellung_hinzufuegen(Bestellung(9001, 120))
kunde.bestellung_hinzufuegen(Bestellung(9002, 35))

len(kunde.bestellungen), [b.bestell_nr for b in kunde.bestellungen]
```

## Kurz-Reflexion
- `Kunde` und `Bestellung` sind Businessobjekte im Code.
- Im ERD sind sie Entitaeten.
- Die Liste `bestellungen` zeigt die 1:n-Beziehung.


## Musterlösung: Klassen und ERD-Zuordnung
Die folgende Implementierung verwendet dieselben IDs und zeigt die 1:n-Beziehungen über Listen bzw. Referenzen.

```python
class Kunde:
    def __init__(self, kunden_id, name): self.kunden_id, self.name, self.rechnungen = kunden_id, name, []
class Artikel:
    def __init__(self, artikel_id, bezeichnung, einzelpreis, mwst_satz): self.artikel_id, self.bezeichnung, self.einzelpreis, self.mwst_satz = artikel_id, bezeichnung, einzelpreis, mwst_satz
class Rechnungsposition:
    def __init__(self, artikel, menge): self.artikel, self.menge = artikel, menge
    @property
    def summe(self): return round(self.menge * self.artikel.einzelpreis, 2)
class Rechnung:
    def __init__(self, rechnung_id, kunde): self.rechnung_id, self.kunde, self.positionen = rechnung_id, kunde, []; kunde.rechnungen.append(self)
    def position_hinzufuegen(self, position): self.positionen.append(position)
    @property
    def netto(self): return round(sum(p.summe for p in self.positionen), 2)

kunde = Kunde(1, "Alpen Technik AG")
kabel = Artikel(1001, "USB-C Kabel", 12.50, 7.7)
buch = Artikel(1002, "Notizbuch Recycling", 8.90, 2.5)
rechnung = Rechnung(1, kunde)
rechnung.position_hinzufuegen(Rechnungsposition(kabel, 2)); rechnung.position_hinzufuegen(Rechnungsposition(buch, 1))
print(kunde.name, rechnung.netto, "CHF", len(rechnung.positionen), "Positionen")
assert rechnung.netto == 33.90
```
