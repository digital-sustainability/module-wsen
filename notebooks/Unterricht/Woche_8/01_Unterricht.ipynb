## 1. Warum OOP? Ein Alltagsvergleich

Stell dir eine Backform vor:
- Die **Form** ist wie eine **Klasse**: Sie gibt vor, wie etwas aussehen soll.
- Ein **Keks** ist wie ein **Objekt**: ein konkretes Exemplar.
- Auf jedem Keks können unterschiedliche Details stehen, zum Beispiel Name oder Farbe.

Genau so arbeiten wir in Python: Ein Bauplan wird einmal definiert, danach können wir viele konkrete Objekte daraus erzeugen.

**Die Kernidee von OOP:** Daten (Attribute) und das, was man mit diesen Daten tut (Methoden), werden an einem Ort gebündelt.

| Ohne OOP | Mit OOP |
|---|---|
| Daten in Variablen/Listen/Dictionaries | Daten als Attribute im Objekt |
| Funktionen stehen lose daneben | Methoden gehören zum Objekt |
| Man muss wissen, welche Funktion zu welchen Daten passt | Das Objekt kennt seine eigenen Daten und sein Verhalten |

## 2. Klasse und Objekt

**Klasse** = Bauplan / Vorlage.
**Objekt** = ein konkretes Exemplar dieser Vorlage (auch **Instanz** genannt).

Wir starten mit einer ganz kleinen Klasse. Noch ohne Attribute und ohne Methoden.


```python
class Kunde:
    pass

kunde_1 = Kunde()
kunde_2 = Kunde()
print(type(kunde_1))
print(kunde_1 is kunde_2)  # False: zwei verschiedene Objekte
```

**Was zeigt `type(kunde_1)`?**

Python bestätigt, dass `kunde_1` ein Objekt der Klasse `Kunde` ist.
Aus einer Klasse können beliebig viele Objekte entstehen. Sie sind unabhängig voneinander (`is` prüft, ob es dasselbe Objekt ist).

> Merke: Klassennamen schreiben wir mit Grossbuchstaben, zum Beispiel `Kunde` oder `Bestellung`.

## 3. Attribute: Daten im Objekt speichern

Ein Objekt wird interessanter, wenn es eigene Daten hält.
Dafür benutzen wir `__init__`.

`__init__` wird automatisch ausgeführt, wenn wir ein neues Objekt erzeugen.
Die Werte, die wir übergeben, speichern wir mit `self.xxx = ...` als Attribute am Objekt.


```python
class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name

kunde = Kunde(1001, "Mia Keller")
print(kunde.kunden_nr)
print(kunde.name)
```

**Mini-Check**

- `Kunde` ist die Klasse.
- `kunde` ist das Objekt.
- `kunden_nr` und `name` sind Attribute.

> Warum schreiben wir `self.name = name`?
> Weil der linke Teil das Attribut am Objekt ist und der rechte Teil der übergebene Wert.


### 3.1 Was passiert bei `Kunde(1001, "Mia Keller")` genau?

`__init__` heisst **Konstruktor** (genauer: Initialisierer). Python führt beim Erzeugen drei Schritte aus:

1. Python legt ein **leeres Objekt** der Klasse `Kunde` an.
2. Python ruft `__init__` auf und übergibt dieses neue Objekt automatisch als erstes Argument `self`.
3. Die restlichen Argumente (`1001`, `"Mia Keller"`) landen in den Parametern `kunden_nr` und `name`. Danach wird das fertige Objekt zurückgegeben.

Darum hat `__init__` einen Parameter mehr, als wir beim Aufruf angeben: `self` setzt Python selbst ein.

**Wichtig:**
- `__init__` hat **kein** `return`. Es richtet das Objekt nur ein.
- Die doppelten Unterstriche (`__init__`) zeigen: Das ist ein spezieller Name, den Python selbst aufruft. Man nennt solche Methoden *dunder methods* (double underscore).
- Wir rufen `__init__` nie direkt auf, sondern schreiben `Kunde(...)`.


```python
class Kunde:
    def __init__(self, kunden_nr, name):
        print("__init__ läuft jetzt. self ist:", self)
        self.kunden_nr = kunden_nr
        self.name = name

kunde = Kunde(1001, "Mia Keller")
print("Fertiges Objekt:  ", kunde)
```

Die beiden Ausgaben zeigen dieselbe Speicheradresse: `self` innerhalb von `__init__` und `kunde` ausserhalb sind **dasselbe Objekt**.

### 3.2 Jedes Objekt hat seine eigenen Daten

`self` sorgt dafür, dass jedes Objekt seine eigenen Attribute besitzt. Ändern wir ein Objekt, bleibt das andere unberührt.


```python
mia = Kunde(1001, "Mia Keller")
ben = Kunde(1002, "Ben Frei")

ben.name = "Ben Frey"   # Attribut nachträglich ändern

print(mia.name)
print(ben.name)
```

### 3.3 Standardwerte und Schlüsselwort-Argumente

Parameter von `__init__` können **Standardwerte** haben. Dann sind sie beim Erzeugen optional.
Man kann Argumente auch **mit Namen** übergeben. Das macht den Aufruf lesbarer.

Attribute dürfen auch Werte haben, die nicht übergeben werden, z.B. eine leere Liste oder ein fixer Startwert.


```python
class Kunde:
    def __init__(self, kunden_nr, name, ort="unbekannt"):
        self.kunden_nr = kunden_nr
        self.name = name
        self.ort = ort
        self.aktiv = True   # fixer Startwert, wird nicht übergeben

k1 = Kunde(1, "Mia Keller", "Bern")
k2 = Kunde(2, "Ben Frey")
k3 = Kunde(name="Lina Roth", kunden_nr=3, ort="Zürich")

print(k1.ort, k2.ort, k3.ort)
print(k2.aktiv)
```

### 3.4 Typische Fehler mit `__init__`

| Fehler | Folge |
|---|---|
| `self` im Kopf vergessen: `def __init__(kunden_nr, name)` | `TypeError` beim Erzeugen |
| `self.` beim Speichern vergessen: `name = name` | Das Attribut wird nicht angelegt, der Wert geht verloren |
| Falsche Anzahl Argumente: `Kunde(1)` | `TypeError: missing 1 required positional argument` |
| `init` statt `__init__` oder nur ein Unterstrich | Python ruft die Methode nicht automatisch auf |

Die Fehlermeldung lesen hilft: Sie sagt, welcher Parameter fehlt (Rückblick Woche 7).


```python
try:
    Kunde(1)   # name fehlt
except TypeError as fehler:
    print("Fehler:", fehler)
```

## 4. Methoden: Verhalten ins Objekt einbauen

Eine Methode gehört zu einem Objekt und beschreibt, was das Objekt tun kann.

Wir erweitern unsere Klasse um eine einfache Methode, die einen Satz zurückgibt.


```python
class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name

    def vorstellen(self):
        return f"Hallo, ich bin {self.name}."

kunde = Kunde(1001, "Mia Keller")
print(kunde.vorstellen())
```

`self` ist das aktuelle Objekt.
Darum rufen wir die Methode mit Punkt-Notation auf: `kunde.vorstellen()`.

> Die Methode gehört zum Objekt, nicht zu einer losen Funktion.

Eine Methode ist also eine Funktion in einer Klasse. Der erste Parameter ist immer `self`. Beim Aufruf `kunde.vorstellen()` setzt Python `kunde` automatisch als `self` ein.


### 4.1 Methoden mit Parametern und Zustandsänderung

Methoden können
- Attribute **lesen** (`self.name`),
- Attribute **verändern** (`self.guthaben += betrag`),
- zusätzliche **Parameter** entgegennehmen und einen Wert **zurückgeben**.

Das Objekt behält seinen Zustand zwischen den Aufrufen.


```python
class Konto:
    def __init__(self, inhaber, guthaben=0):
        self.inhaber = inhaber
        self.guthaben = guthaben

    def einzahlen(self, betrag):
        self.guthaben += betrag

    def abheben(self, betrag):
        if betrag > self.guthaben:
            return False   # nicht genug Guthaben
        self.guthaben -= betrag
        return True

konto = Konto("Mia Keller", 100)
konto.einzahlen(50)
print(konto.guthaben)                 # 150
print(konto.abheben(200), konto.guthaben)  # False 150
print(konto.abheben(30), konto.guthaben)   # True 120
```

Eine Methode kann auch andere Methoden desselben Objekts aufrufen, immer über `self.`:
`self.einzahlen(10)`. Ohne `self.` sucht Python eine lose Funktion und findet sie nicht (`NameError`).

### 4.2 Attribute vs. lokale Variablen

| | Schreibweise | Lebensdauer |
|---|---|---|
| Attribut | `self.guthaben` | solange das Objekt lebt |
| Parameter / lokale Variable | `betrag` | nur während des Methodenaufrufs |


### 4.3 Lesbare Ausgabe mit `__str__`

Wenn wir ein Objekt direkt ausgeben, zeigt Python nur eine unlesbare Adresse (wie vorher bei `print(kunde)`).
Mit der speziellen Methode `__str__` bestimmen wir selbst, was `print(objekt)` ausgibt. Sie muss einen Text **zurückgeben**.


```python
class Konto:
    def __init__(self, inhaber, guthaben=0):
        self.inhaber = inhaber
        self.guthaben = guthaben

    def __str__(self):
        return f"Konto von {self.inhaber}: CHF {self.guthaben:.2f}"

print(Konto("Mia Keller", 120))
```

## 5. Anwendung im Projekt: Klassen für das Rechnungsmodell
Auch im Rechnungsprojekt arbeiten wir in SW8 mit **Klassen, Attributen und Methoden**. `Kunde`, `Artikel`, `Rechnung` und `Rechnungsposition` werden als Objekte modelliert; die fachlichen Beziehungen laufen über Referenzen und Listen - **nicht** über Dictionaries. Eine Positionssumme berechnen wir mit der normalen Methode `summe()`, die Rechnung addiert diese Werte mit `netto()`.


```python
class Kunde:
    def __init__(self, kunden_id, name):
        self.kunden_id = kunden_id
        self.name = name
        self.rechnungen = []

class Artikel:
    def __init__(self, artikel_id, bezeichnung, einzelpreis, mwst_satz):
        self.artikel_id = artikel_id
        self.bezeichnung = bezeichnung
        self.einzelpreis = einzelpreis
        self.mwst_satz = mwst_satz

class Rechnungsposition:
    def __init__(self, artikel, menge):
        self.artikel = artikel
        self.menge = menge

    def summe(self):
        return round(self.menge * self.artikel.einzelpreis, 2)

class Rechnung:
    def __init__(self, rechnung_id, kunde):
        self.rechnung_id = rechnung_id
        self.kunde = kunde
        self.positionen = []
        kunde.rechnungen.append(self)

    def position_hinzufuegen(self, position):
        self.positionen.append(position)

    def netto(self):
        total = 0
        for position in self.positionen:
            total += position.summe()
        return round(total, 2)

kunde = Kunde(1, "Alpen Technik AG")
kabel = Artikel(1001, "USB-C Kabel", 12.50, 7.7)
buch = Artikel(1002, "Notizbuch Recycling", 8.90, 2.5)
rechnung = Rechnung(1, kunde)
rechnung.position_hinzufuegen(Rechnungsposition(kabel, 2))
rechnung.position_hinzufuegen(Rechnungsposition(buch, 1))

print(kunde.name, rechnung.netto(), "CHF", len(rechnung.positionen), "Positionen")
assert rechnung.netto() == 33.90
```

## 6. Vom Code zum Businessobjekt und zum ERD

Ein **Businessobjekt** ist ein Ding aus der fachlichen Welt, zum Beispiel `Kunde` oder `Bestellung`.
Ein **ERD** zeigt diese Dinge als Entitäten und ihre Beziehungen.

Wenn ein Kunde mehrere Bestellungen haben kann, ist das im ERD eine **1:n-Beziehung**.
Im Code speichern wir diese Beziehung oft mit einer Liste.


```python
class Bestellung:
    def __init__(self, bestell_nr, betrag):
        self.bestell_nr = bestell_nr
        self.betrag = betrag

class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name
        self.bestellungen = []

    def bestellung_hinzufuegen(self, bestellung):
        self.bestellungen.append(bestellung)

kunde = Kunde(7, "Lina")
kunde.bestellung_hinzufuegen(Bestellung(9001, 120))
kunde.bestellung_hinzufuegen(Bestellung(9002, 35))

print(len(kunde.bestellungen))
print(kunde.bestellungen[0].bestell_nr)
```

**Was bedeutet das?**

- `Kunde` und `Bestellung` sind Businessobjekte.
- Im ERD sind sie Entitäten.
- Die Liste `bestellungen` zeigt die 1:n-Beziehung im Code.
