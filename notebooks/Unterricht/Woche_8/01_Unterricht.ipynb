## 1. Warum OOP? Ein Alltagsvergleich

Stell dir eine Backform vor:
- Die **Form** ist wie eine **Klasse**: Sie gibt vor, wie etwas aussehen soll.
- Ein **Keks** ist wie ein **Objekt**: ein konkretes Exemplar.
- Auf jedem Keks koennen unterschiedliche Details stehen, zum Beispiel Name oder Farbe.

Genau so arbeiten wir in Python: Ein Bauplan wird einmal definiert, danach koennen wir viele konkrete Objekte daraus erzeugen.


---
## 2. Klasse und Objekt

**Klasse** = Bauplan / Vorlage.
**Objekt** = ein konkretes Exemplar dieser Vorlage.

Wir starten mit einer ganz kleinen Klasse. Noch ohne Attribute und ohne Methoden.


```python
class Kunde:
    pass

kunde_1 = Kunde()
print(type(kunde_1))
```

**Was zeigt `type(kunde_1)`?**

Python bestaetigt, dass `kunde_1` ein Objekt der Klasse `Kunde` ist.

> Merke: Klassennamen schreiben wir mit Grossbuchstaben, zum Beispiel `Kunde` oder `Bestellung`.


---
## 3. Attribute: Daten im Objekt speichern

Ein Objekt wird interessanter, wenn es eigene Daten haelt.
Dafuer benutzen wir `__init__`.

`__init__` wird automatisch ausgefuehrt, wenn wir ein neues Objekt erzeugen.
Die Werte, die wir uebergeben, speichern wir mit `self.xxx = ...` als Attribute am Objekt.


```python
class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name

kunde = Kunde(1001, "Mia Keller")
print(kunde.kunden_nr)
print(kunde.name)
```

**Mini-Check**

- `Kunde` ist die Klasse.
- `kunde` ist das Objekt.
- `kunden_nr` und `name` sind Attribute.

> Warum schreiben wir `self.name = name`?
> Weil der linke Teil das Attribut am Objekt ist und der rechte Teil der uebergebene Wert.


---
## 4. Methoden: Verhalten ins Objekt einbauen

Eine Methode gehoert zu einem Objekt und beschreibt, was das Objekt tun kann.

Wir erweitern unsere Klasse um eine einfache Methode, die einen Satz zurueckgibt.


```python
class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name

    def vorstellen(self):
        return f"Hallo, ich bin {self.name}."

kunde = Kunde(1001, "Mia Keller")
print(kunde.vorstellen())
```

`self` ist das aktuelle Objekt.
Darum rufen wir die Methode mit Punkt-Notation auf: `kunde.vorstellen()`.

> Die Methode gehoert zum Objekt, nicht zu einer losen Funktion.

## 5. Vom Code zum Businessobjekt und zum ERD

Ein **Businessobjekt** ist ein Ding aus der fachlichen Welt, zum Beispiel `Kunde` oder `Bestellung`.
Ein **ERD** zeigt diese Dinge als Entitaeten und ihre Beziehungen.

Wenn ein Kunde mehrere Bestellungen haben kann, ist das im ERD eine **1:n-Beziehung**.
Im Code speichern wir diese Beziehung oft mit einer Liste.


```python
class Bestellung:
    def __init__(self, bestell_nr, betrag):
        self.bestell_nr = bestell_nr
        self.betrag = betrag

class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name
        self.bestellungen = []

    def bestellung_hinzufuegen(self, bestellung):
        self.bestellungen.append(bestellung)

kunde = Kunde(7, "Lina")
kunde.bestellung_hinzufuegen(Bestellung(9001, 120))
kunde.bestellung_hinzufuegen(Bestellung(9002, 35))

print(len(kunde.bestellungen))
print(kunde.bestellungen[0].bestell_nr)
```

**Was bedeutet das?**

- `Kunde` und `Bestellung` sind Businessobjekte.
- Im ERD sind sie Entitaeten.
- Die Liste `bestellungen` zeigt die 1:n-Beziehung im Code.

## Zusammenfassung
- Die **Klasse** ist der Bauplan.
- Das **Objekt** ist das konkrete Exemplar.
- Ein **Attribut** speichert Daten.
- Eine **Methode** beschreibt Verhalten.
- Businessobjekte im Code koennen wir direkt mit ERD-Entitaeten verbinden.

## Anwendung im Projekt: Fachmodell und ERD
Die Klassen bilden die ERD-Entitäten ab: **Kunde 1:n Rechnung**, **Rechnung 1:n Rechnungsposition** und **Artikel 1:n Rechnungsposition**. `Rechnung` besitzt Positionen; eine Position referenziert genau einen Artikel. IDs entsprechen den SQLite-Schlüsseln aus SW5.

```python
class Kunde:
    def __init__(self, kunden_id, name): self.kunden_id, self.name, self.rechnungen = kunden_id, name, []
class Artikel:
    def __init__(self, artikel_id, bezeichnung, einzelpreis, mwst_satz): self.artikel_id, self.bezeichnung, self.einzelpreis, self.mwst_satz = artikel_id, bezeichnung, einzelpreis, mwst_satz
class Rechnungsposition:
    def __init__(self, artikel, menge): self.artikel, self.menge = artikel, menge
    @property
    def summe(self): return round(self.menge * self.artikel.einzelpreis, 2)
class Rechnung:
    def __init__(self, rechnung_id, kunde): self.rechnung_id, self.kunde, self.positionen = rechnung_id, kunde, []; kunde.rechnungen.append(self)
    def position_hinzufuegen(self, position): self.positionen.append(position)
    @property
    def netto(self): return round(sum(p.summe for p in self.positionen), 2)

kunde = Kunde(1, "Alpen Technik AG")
kabel = Artikel(1001, "USB-C Kabel", 12.50, 7.7)
buch = Artikel(1002, "Notizbuch Recycling", 8.90, 2.5)
rechnung = Rechnung(1, kunde)
rechnung.position_hinzufuegen(Rechnungsposition(kabel, 2)); rechnung.position_hinzufuegen(Rechnungsposition(buch, 1))
print(kunde.name, rechnung.netto, "CHF", len(rechnung.positionen), "Positionen")
assert rechnung.netto == 33.90
```
