## 1. Warum OOP? Ein Alltagsvergleich

Stell dir eine Backform vor:
- Die **Form** ist wie eine **Klasse**: Sie gibt vor, wie etwas aussehen soll.
- Ein **Keks** ist wie ein **Objekt**: ein konkretes Exemplar.
- Auf jedem Keks können unterschiedliche Details stehen, zum Beispiel Name oder Farbe.

Genau so arbeiten wir in Python: Ein Bauplan wird einmal definiert, danach können wir viele konkrete Objekte daraus erzeugen.


---
## 2. Klasse und Objekt

**Klasse** = Bauplan / Vorlage.
**Objekt** = ein konkretes Exemplar dieser Vorlage.

Wir starten mit einer ganz kleinen Klasse. Noch ohne Attribute und ohne Methoden.


```python
class Kunde:
    pass

kunde_1 = Kunde()
print(type(kunde_1))
```

**Was zeigt `type(kunde_1)`?**

Python bestätigt, dass `kunde_1` ein Objekt der Klasse `Kunde` ist.

> Merke: Klassennamen schreiben wir mit Grossbuchstaben, zum Beispiel `Kunde` oder `Bestellung`.


---
## 3. Attribute: Daten im Objekt speichern

Ein Objekt wird interessanter, wenn es eigene Daten hält.
Dafür benutzen wir `__init__`.

`__init__` wird automatisch ausgeführt, wenn wir ein neues Objekt erzeugen.
Die Werte, die wir übergeben, speichern wir mit `self.xxx = ...` als Attribute am Objekt.


```python
class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name

kunde = Kunde(1001, "Mia Keller")
print(kunde.kunden_nr)
print(kunde.name)
```

**Mini-Check**

- `Kunde` ist die Klasse.
- `kunde` ist das Objekt.
- `kunden_nr` und `name` sind Attribute.

> Warum schreiben wir `self.name = name`?
> Weil der linke Teil das Attribut am Objekt ist und der rechte Teil der übergebene Wert.

## 4. Methoden: Verhalten ins Objekt einbauen

Eine Methode gehört zu einem Objekt und beschreibt, was das Objekt tun kann.

Wir erweitern unsere Klasse um eine einfache Methode, die einen Satz zurückgibt.


```python
class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name

    def vorstellen(self):
        return f"Hallo, ich bin {self.name}."

kunde = Kunde(1001, "Mia Keller")
print(kunde.vorstellen())
```

`self` ist das aktuelle Objekt.
Darum rufen wir die Methode mit Punkt-Notation auf: `kunde.vorstellen()`.

> Die Methode gehört zum Objekt, nicht zu einer losen Funktion.


## Anwendung im Projekt: Klassen für das Rechnungsmodell
Auch im Rechnungsprojekt arbeiten wir in SW8 mit **Klassen, Attributen und Methoden**. `Kunde`, `Artikel`, `Rechnung` und `Rechnungsposition` werden als Objekte modelliert; die fachlichen Beziehungen laufen über Referenzen und Listen - **nicht** über Dictionaries. Eine Positionssumme berechnen wir mit der normalen Methode `summe()`, die Rechnung addiert diese Werte mit `netto()`.


```python
class Kunde:
    def __init__(self, kunden_id, name):
        self.kunden_id = kunden_id
        self.name = name
        self.rechnungen = []

class Artikel:
    def __init__(self, artikel_id, bezeichnung, einzelpreis, mwst_satz):
        self.artikel_id = artikel_id
        self.bezeichnung = bezeichnung
        self.einzelpreis = einzelpreis
        self.mwst_satz = mwst_satz

class Rechnungsposition:
    def __init__(self, artikel, menge):
        self.artikel = artikel
        self.menge = menge

    def summe(self):
        return round(self.menge * self.artikel.einzelpreis, 2)

class Rechnung:
    def __init__(self, rechnung_id, kunde):
        self.rechnung_id = rechnung_id
        self.kunde = kunde
        self.positionen = []
        kunde.rechnungen.append(self)

    def position_hinzufuegen(self, position):
        self.positionen.append(position)

    def netto(self):
        total = 0
        for position in self.positionen:
            total += position.summe()
        return round(total, 2)

kunde = Kunde(1, "Alpen Technik AG")
kabel = Artikel(1001, "USB-C Kabel", 12.50, 7.7)
buch = Artikel(1002, "Notizbuch Recycling", 8.90, 2.5)
rechnung = Rechnung(1, kunde)
rechnung.position_hinzufuegen(Rechnungsposition(kabel, 2))
rechnung.position_hinzufuegen(Rechnungsposition(buch, 1))

print(kunde.name, rechnung.netto(), "CHF", len(rechnung.positionen), "Positionen")
assert rechnung.netto() == 33.90
```

## 5. Vom Code zum Businessobjekt und zum ERD

Ein **Businessobjekt** ist ein Ding aus der fachlichen Welt, zum Beispiel `Kunde` oder `Bestellung`.
Ein **ERD** zeigt diese Dinge als Entitäten und ihre Beziehungen.

Wenn ein Kunde mehrere Bestellungen haben kann, ist das im ERD eine **1:n-Beziehung**.
Im Code speichern wir diese Beziehung oft mit einer Liste.


```python
class Bestellung:
    def __init__(self, bestell_nr, betrag):
        self.bestell_nr = bestell_nr
        self.betrag = betrag

class Kunde:
    def __init__(self, kunden_nr, name):
        self.kunden_nr = kunden_nr
        self.name = name
        self.bestellungen = []

    def bestellung_hinzufuegen(self, bestellung):
        self.bestellungen.append(bestellung)

kunde = Kunde(7, "Lina")
kunde.bestellung_hinzufuegen(Bestellung(9001, 120))
kunde.bestellung_hinzufuegen(Bestellung(9002, 35))

print(len(kunde.bestellungen))
print(kunde.bestellungen[0].bestell_nr)
```

**Was bedeutet das?**

- `Kunde` und `Bestellung` sind Businessobjekte.
- Im ERD sind sie Entitäten.
- Die Liste `bestellungen` zeigt die 1:n-Beziehung im Code.
