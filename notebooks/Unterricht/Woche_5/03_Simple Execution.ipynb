## Manipulation der Datenbank via einfache SQL-Statements

Für Manipulationen der DB ist ein sog. **Cursor**-Objekt erforderlich:

```python
import sqlite3

con=sqlite3.connect("example.db")
cursorObject = con.cursor()
```

Mit der Methode **execute()** dieses Objektes kann jetzt ein SQL-Befehl ausgeführt werden, z.B. zum Erstellen einer neuen Tabelle:


```python
cursorObject.execute('''CREATE TABLE Mitarbeiter (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    alt INTEGER NOT NULL,
    adresse VARCHAR(30) NOT NULL,
    lohn REAL,
    verheiratet BOOLEAN,
    departement TEXT)''')
```

Oder auch Werte in die Tabelle eingefügt werden:

```python
cursorObject.execute('''INSERT INTO Mitarbeiter VALUES (1, "Anja",27,"Zurich",10000,"ja","wirtschaft")''')
cursorObject.execute('''INSERT INTO Mitarbeiter VALUES (2, "Flurina",29,"Bern",7000,"nein","kunst")''')
cursorObject.execute('''INSERT INTO Mitarbeiter VALUES (3, "Michael",25,"Gerzensee",50000,"nein","IT")''')
cursorObject.execute('''INSERT INTO Mitarbeiter VALUES (4, "Amir",31,"Thun",9000,"ja","wirtschaft")''')
cursorObject.execute('''INSERT INTO Mitarbeiter VALUES (5, "Maria",23,"Bern",3000,"nein","IT")''')
cursorObject.execute('''INSERT INTO Mitarbeiter VALUES (6, "Urs",44,"Biel",10000,"ja","gesundheit")''')
cursorObject.execute('''INSERT INTO Mitarbeiter VALUES (7, "Kathrin",37,"Vevey",6000,"ja","wirtschaft")''')

con.commit()
```

Die Methode **commit()** stellt sicher, dass allenfalls offene Transaktionen abgeschlossen und alle Änderungen in der Datenbank auch wirklich abgespeichert werden.

Die hier benutzten Python-Objekte für die Verbindung (_con_ im Beispiel oben) sowie den Cursor verfügen über viele weitere Methoden. Diese sind in der [sqlite3 — DB-API 2.0 interface for SQLite databases](https://docs.python.org/3/library/sqlite3.html) dokumentiert.
