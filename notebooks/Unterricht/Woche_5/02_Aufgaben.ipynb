### Aufgabe 3.2 (Schritt 2 – Etwas schwerer)**Aufgabe:**1. Schreib eine SQL-Abfrage, die alle Produkte mit Preis > 50€ zeigt.2. Sortiere sie nach Preis absteigend.3. Gib die Ergebnisse aus.**Tipp:** Nutze `SELECT`, `WHERE`, und `ORDER BY`.

### Aufgabe 3.3 (Schritt 3 – Transfer)**Aufgabe:**1. Schreib eine SQL-Abfrage, die die Durchschnittspreise pro Kategorie berechnet.2. Konvertiere das Resultat in einen DataFrame.3. Gib aus, welche Kategorie am teuersten ist.**Tipp:** Nutze `GROUP BY` und `AVG()` in SQL.

````python
# Deine Lösung:cur.execute("""SELECT kategorie, AVG(preis) as durchschnitt_preis FROM produkte GROUP BY kategorie ORDER BY durchschnitt_preis DESC""")durchschnitte = cur.fetchall()print("Durchschnittspreise pro Kategorie:")for row in durchschnitte:    print(f"  {row[0]}: {row[1]:.2f}€")# In DataFrame konvertierendf_durchschnitte = pd.DataFrame(durchschnitte, columns=["kategorie", "durchschnitt_preis"])print(f"\nTeuerste Kategorie: {df_durchschnitte.iloc[0]['kategorie']}")
````
