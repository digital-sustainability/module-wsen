## 1. Listen von Dictionaries – Die Basis

### Was ist eine "Liste von Dictionaries"?

Eine **Liste von Dictionaries** ist die perfekte Darstellung für tabellarische Daten:
- Jedes Element in der Liste ist ein Dictionary (eine Zeile mit Key-Value-Paaren)
- Alle Dictionaries haben die gleichen Keys (Spaltennamen)
- Jeder Key wird später zu einer Spalte in einer Tabelle

Beispiel: Eine Klassenliste als Liste von Dictionaries:
```
[
  {"id": 1, "name": "Anna", "punkte": 84},    ← Zeile 1
  {"id": 2, "name": "Ben", "punkte": 71},     ← Zeile 2
  {"id": 3, "name": "Clara", "punkte": 93},   ← Zeile 3
]
```

````python
import sqlite3
import pandas as pd
````

### Arbeitsbeispiel: Eine einfache Klassenliste

## 2. Pandas DataFrames – Schnelle tabellarische Operationen

### Was ist ein Pandas DataFrame?

Ein **DataFrame** ist wie eine Excel-Tabelle in Python:
- Spalten mit Namen (wie `name`, `punkte`)
- Zeilen mit Indizes (0, 1, 2, ...)
- Effiziente Operationen: filtern, berechnen, statistiken

Pandas ist viel schneller als manuelle Schleifen und viel lesbar.

### Von Liste zu DataFrame

Die Konvertierung ist sehr einfach – Pandas erkennt die Struktur automatisch:

````python
# Konvertiere die Liste in einen DataFrame
df_students = pd.DataFrame(students)

# Anschauen:
print(df_students)
````

### Den DataFrame verstehen

Drei wichtige Eigenschaften eines DataFrames:

````python
# 1. shape: (Anzahl Zeilen, Anzahl Spalten)
print(f"Form: {df_students.shape}")
print(f"Das bedeutet: {df_students.shape[0]} Zeilen, {df_students.shape[1]} Spalten")

# 2. columns: Die Spaltennamen
print(f"\nSpalten: {list(df_students.columns)}")

# 3. dtypes: Die Datentypen der Spalten
print(f"\nDatentypen:")
print(df_students.dtypes)
````

### Filtern in Pandas (wie Bedingungen in Woche 3)

In Woche 3 habt ihr `if`-Bedingungen geschrieben. In Pandas funktioniert das so ähnlich – nur für **ganze Spalten** auf einmal:

````python
# Bedingung: punkte >= 70
# Pandas erzeugt eine Maske (True/False für jede Zeile)
maske = df_students["punkte"] >= 70
print("Maske (True/False pro Zeile):")
print(maske)

# Nutze die Maske um zu filtern:
bestanden = df_students[maske]
print("\nStudenten mit 70+ Punkten:")
print(bestanden)
````

### Warum funktioniert das?

Der Ausdruck `df_students["punkte"] >= 70` erzeugt für **jede Zeile** einen `True`/`False`-Wert.
Diese Werte bilden eine "Maske" – nur die Zeilen mit `True` werden behalten.

### Vergleich: Schleife vs. Pandas

Beide Wege geben das gleiche Resultat, Pandas ist aber kurz und schnell:

**Mit Schleife (Woche 4):**
```python
bestanden_namen = []
for s in students:
    if s["punkte"] >= 70:
        bestanden_namen.append(s["name"])
```

**Mit Pandas (heute):**
```python
bestanden = df_students[df_students["punkte"] >= 70]
```

### Einfache Statistiken mit Pandas

Pandas hat eingebaute Funktionen für häufig verwendete Berechnungen:

````python
# Statistiken über die Spalte 'punkte'
print(f"Durchschnitt: {df_students['punkte'].mean():.2f}")
print(f"Minimum: {df_students['punkte'].min()}")
print(f"Maximum: {df_students['punkte'].max()}")
print(f"Summe: {df_students['punkte'].sum()}")
print(f"Standardabweichung: {df_students['punkte'].std():.2f}")
````

### Einfache Visualisierungen

Mithilfe von Pandas können wir auch einfache Visualisierungen erstellen:


````python
import seaborn as sns

sns.barplot(data=df_students, x="name", y="punkte")
````

`seaborn` ist eine Python-Bibliothek zur Datenvisualisierung, die auf `matplotlib` basiert. Sie bietet eine benutzerfreundliche Schnittstelle zum Erstellen ansprechender und aussagekräftiger statistischer Grafiken. Es gibt auch ausführliche Dokumentation und Tutorials auf der [offiziellen Website](https://seaborn.pydata.org).

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Zusatzinformation für eine aussagekräftigere Visualisierung
df_students["status"] = df_students["punkte"].apply(lambda x: "bestanden" if x >= 70 else "nicht bestanden")

sns.set_style("whitegrid")
fig, ax = plt.subplots(figsize=(10, 6))

# Komplexere Visualisierung:
# - x-Achse: Student-ID
# - y-Achse: Punkte
# - Farbe: bestanden / nicht bestanden
# - Größe: Punktewert
sns.scatterplot(
    data=df_students,
    x="id",
    y="punkte",
    hue="status",
    size="punkte",
    sizes=(100, 500),
    palette={"bestanden": "#2E8B57", "nicht bestanden": "#D9534F"},
    alpha=0.8,
    ax=ax
)

# Trendlinie ergänzen
sns.lineplot(
    data=df_students.sort_values("id"),
    x="id",
    y="punkte",
    color="gray",
    linestyle="--",
    linewidth=1.5,
    ax=ax,
    label="Punkteverlauf"
)

# Namen der Studenten über den Punkten anzeigen
for _, row in df_students.iterrows():
    ax.text(row["id"], row["punkte"] + 2, row["name"], ha="center", va="bottom", fontsize=9)

ax.set_title("Studentenleistung im Vergleich")
ax.set_xlabel("Studenten-ID")
ax.set_ylabel("Punkte")
ax.legend(title="Status", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
````

Der Datenrahmen kann jetzt zur Analyse der Daten genutzt werden, ohne dass weitere Abfragen in der Datenbank notwendig sind. Zum Beispiel, um folgende Frage beantworten zu können:

_"Finden Sie die weltweiten Verkaufszahlen des Videospiels mit dem Namen "Minecraft"."_

In einem ersten Schritt erzeugen wir eine Liste, welche für jede Zeile des Rahmens angibt, ob diese in der Spalte "Name" den Wert "Minecraft" stehen hat.

````python
import pandas as pd

videogames_df = pd.read_csv("vgsales.csv")

minecraft_idx = videogames_df["Name"] == "Minecraft"
minecraft_idx
````

Diese kann jetzt als Selektor genutzt werden, um die relevanten Zeilen (mit allen Spalten) herauszuholen:

````python
minecraft_rows = videogames_df[minecraft_idx]
minecraft_rows
````

Und zum Schluss bilden wir noch die Summe über alle Plattformen hinweg:

````python
minecraft_sales_global = minecraft_rows["Global_Sales"].sum()

print(f"\nInsgesamt wurden {minecraft_sales_global} Millionen Einheiten des Spiels Minecraft verkauft.")
````

Anstatt das Ganze Schritt für Schritt herzuleiten, lässt es sich auch in einer Zeile formulieren

````python
minecraft_sales = videogames_df[videogames_df["Name"] == "Minecraft"]["Global_Sales"].sum()
minecraft_sales
````

Diese Information hätte man natürlich auch mittels eines einzigen SELECT-Befehls erhalten können. Wie würde dieser aussehen? Schreiben Sie den Python-Code dafür:

````python
import sqlite3

# connect to SQLite database
conn = sqlite3.connect('example.db')
cursorObject=conn.cursor()

cursorObject.execute("""
        SELECT SUM(Global_Sales) FROM Videogame_Sales WHERE Name='Minecraft'
    """)
rows=cursorObject.fetchall()

print(f"\nInsgesamt wurden {rows[0][0]} Millionen Einheiten des Spiels Minecraft verkauft.")
    
conn.close()
````
