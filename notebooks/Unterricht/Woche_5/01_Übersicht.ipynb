## Datenbankprogrammierung mit Python - Übersicht

Zur Erinnerung, und wie in der Abbildung unten dargestellt, bestehen Softwaresysteme im Geschäftsumfeld typischerweise aus mind. 3 Schichten (engl. **3-tier architecture**).


<font size="1">_Bildquelle: https://managementmania.com/en/three-tier-architecture_</font>


| _**Schicht**_ | _**Funktion**_ | _**Illustration**_ |
| :- | :- | :- |
| **Presentation&nbsp;Tier** | Die Präsentationsschicht ist für die _Interaktion mit dem Benutzer oder der Benutzerin_ verantwortlich. Dazu gehören die Darstellung der Informationen sowie die Möglichkeiten, Daten einzugeben. Wird oft auch als _Front-End_ bezeichnet.<br><br>Beispiele: <ul><li>Grafische Benutzeroberfläche (GUI; Graphical User Interface)</li><li>Webseite z.B einer E-Commerce Anwendung</li></ul>| 
| **Application Tier** | Die Anwendungsschicht ist für die Datenverarbeitung und die _Geschäftslogik_ verantwortlich. Als Schicht zwischen Präsentations- und Datenschicht bereitet sie die Daten für die Präsentation auf, verarbeitet Nutzereingaben und speichert, löscht und ändert Daten in der Datenschicht. Kann in unterschiedlichen Programmiersprachen geschrieben sein: C++, C#, Java, Python, ABAP etc. Bildet zusammen mit der Datenschicht das _Back-End_.| 
| **Data Tier** | Die Datenschicht ist für die _persistente und konsistente Speicherung der Daten_ verantwortlich. | 



In der heutigen Lektion beschäftigen wir uns mit Python-Anwendungen im _Application Tier_. Allerdings der Einfachheit halber in Jupyter-Notebooks und nicht in einem vollständigen Programm.
