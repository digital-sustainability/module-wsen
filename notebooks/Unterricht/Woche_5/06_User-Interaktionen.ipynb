## Interaktionen mit der DB basierend auf User-Input

In den vorherigen Beispielen wurde ein statischer String jeweils der **execute()**-Methode mitgegeben. Typischerweise ist eine solche harte Kodierung in Programmen nicht sinnvoll; sondern je nach Businesslogik und Kontext wird der passende Befehlsstring dynamisch erzeugt.

Einfaches Beispiel: Abfrage der Daten basierend auf User-Input. So soll der Benutzer z.B. angeben können, an welchen Daten er oder sie interessiert ist. Im folgenden Programm wir der User gebeten, für Namen und Departement ein Filtermuster einzugeben, und es werden nur die Datensätze zurückgeliefert, welche auf das Muster passen. Es können bei der Definition der Muster die SQL-Wildcards '%' und '\_' verwendet werden. Es werden nur Name, Adresse und Department angezeigt, sensitive Daten wie Alter und Lohn nicht:

```python
import sqlite3
from IPython.display import display_markdown

con = sqlite3.connect('example.db')
cursorObject=con.cursor()

name_filter = input(
    "Welche Mitarbeiter sollen zurückgeliefert werden (Wildcards erlaubt): ")
dept_filter = input(
    "Aus welchen Abteilungen (Wildcards erlaubt): ")


sqlQuery=f"SELECT name, adresse, departement FROM Mitarbeiter \
    WHERE name LIKE '{name_filter}' AND departement LIKE '{dept_filter}'"
results=cursorObject.execute(sqlQuery)

rows=results.fetchall()


# Formatierung des Results als Markdown
markdown = '|'
line = '|'

for col in results.description:   
    markdown += col[0] + ' | ' 
    line += '---|'
    
markdown += '\n' + line

for row in rows:
    markdown += f'\n|'
    for i in range(len(results.description)):
        markdown += f' {row[i]} |'

display_markdown(markdown, raw=True)
    
con.close()
```
