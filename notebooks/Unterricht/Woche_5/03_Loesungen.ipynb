---## Lösungen zu Aufgabe 1 – Listen von Dictionaries

### Lösung 1.1: Liste erstellen**Was ist die Idee?**- Jedes Produkt hat gleiche Keys (id, name, preis, kategorie)- Die Keys werden später zu Spaltennamen- Die Reihenfolge der Dictionaries ist beliebig

```python
products = [    {"id": 1, "name": "Laptop", "preis": 899.99, "kategorie": "Elektronik"},    {"id": 2, "name": "Maus", "preis": 25.50, "kategorie": "Elektronik"},    {"id": 3, "name": "T-Shirt", "preis": 19.99, "kategorie": "Kleidung"},    {"id": 4, "name": "Jeans", "preis": 59.99, "kategorie": "Kleidung"},    {"id": 5, "name": "Tisch", "preis": 299.99, "kategorie": "Möbel"},]print("Unsere Produktliste:")print(products)print(f"\nTyp der gesamten Liste: {type(products)}")print(f"Typ des ersten Elements: {type(products[0])}")
```

### Lösung 1.2: Analyse der Liste**Was ist die Idee?**- Mit `len()` zählen wir die Elemente- Mit Schleifen und Vergleichen (wie in Woche 4) iterieren wir durch die Daten- Mit `if`-Bedingungen (wie in Woche 3) filtern wir Datensätze

```python
# Frage 1: Wie viele Produkte?print("FRAGE 1: Anzahl Produkte")print(f"Antwort: {len(products)} Produkte")# Frage 2: Teuerstes Produktprint("\nFRAGE 2: Teuerstes Produkt")max_price = 0teuerstes = Nonefor p in products:    # Für jedes Produkt prüfen: ist es teurer als bisher gefundenes?    if p["preis"] > max_price:        max_price = p["preis"]        teuerstes = pprint(f"Antwort: {teuerstes['name']} für {max_price}€")print(f"Warum? Wir vergleichen jeden Preis, speichern das Maximum ab.")# Frage 3: Wie viele über 50€?print("\nFRAGE 3: Anzahl Produkte über 50€")teuer_count = 0for p in products:    if p["preis"] > 50:        teuer_count += 1print(f"Antwort: {teuer_count} Produkte kosten über 50€")print(f"Warum? Wir zählen nur Produkte, bei denen die Bedingung 'preis > 50' wahr ist.")
```

### Lösung 1.3: Filtern nach Kategorie**Was ist die Idee?**- Mit `if` testen wir die Bedingung- Mit `.append()` fügen wir passende Produkte einer neuen Liste hinzu- Das Ergebnis ist eine Untermenge der Original-Liste

```python
kategorie = "Elektronik"kategorie_elektronik = []for p in products:    # Nur Produkte mit kategorie == "Elektronik" hinzufügen    if p["kategorie"] == kategorie:        kategorie_elektronik.append(p)print(f"Alle Produkte in Kategorie '{kategorie}':")for p in kategorie_elektronik:    print(f"  - {p['name']}: {p['preis']}€")print(f"\nWarum funktioniert das?")print(f"1. Wir iterieren über alle Produkte (for-Schleife)")print(f"2. Für jedes Produkt testen wir die Bedingung (if)")print(f"3. Nur wenn die Bedingung wahr ist, fügen wir es hinzu (append)")print(f"4. Das Ergebnis: nur die Elektronik-Produkte")
```

---## Lösungen zu Aufgabe 2 – Pandas DataFrames

### Lösung 2.1: DataFrame erstellen und inspizieren**Was ist die Idee?**- Ein DataFrame ist wie eine Excel-Tabelle- Mit `shape` sehen wir (Zeilen, Spalten)- Mit `columns` sehen wir die Spaltennamen- Mit `dtypes` sehen wir, wie Python die Daten erkannt hat

```python
import pandas as pd# Konvertiere Liste zu DataFramedf_products = pd.DataFrame(products)print("Unser DataFrame:")print(df_products)# Strukturinformationenprint(f"\nShape (Zeilen, Spalten): {df_products.shape}")print(f"Spaltennamen: {list(df_products.columns)}")print(f"\nDatentypen (wichtig für SQL später!):")print(df_products.dtypes)print(f"\nWarum wichtig?")print(f"- 'id' ist int64 (Ganzzahl) → wird INTEGER in SQL")print(f"- 'name' und 'kategorie' sind object (Text) → wird TEXT in SQL")print(f"- 'preis' ist float64 (Dezimalzahl) → wird REAL in SQL")
```

### Lösung 2.2: Statistiken und Filterung**Was ist die Idee?**- Pandas hat eingebaute Funktionen für Statistiken: `.mean()`, `.min()`, `.max()`- Filterung funktioniert wie Bedingungen in Python, aber auf der ganzen Spalte auf einmal- Das ist viel schneller als mit Schleifen!

```python
# Statistikenprint("STATISTIKEN ÜBER PREISE:")print(f"Durchschnitt: {df_products['preis'].mean():.2f}€")print(f"Minimum: {df_products['preis'].min()}€")print(f"Maximum: {df_products['preis'].max()}€")# Filterung: Alle Produkte > 100€print("\nPRODUKTE ÜBER 100€:")teuer = df_products[df_products['preis'] > 100]print(teuer)print("\nWarum funktioniert die Filterung?")print("Der Ausdruck 'df_products[\"preis\"] > 100' erzeugt:")print(df_products['preis'] > 100)print("Diese True/False-Werte bilden eine 'Maske' – nur True-Zeilen werden behalten!")
```

### Lösung 2.3: Nach Kategorie filtern und sortieren**Was ist die Idee?**- Filterung mit `==` für Text (exakte Übereinstimmung)- Sortierung mit `.sort_values()`- Die Operationen sind verkettbar (nacheinander ausführbar)

```python
kategorie = "Elektronik"# Filtere nach Kategorie UND sortiere nach Preisgefiltert = df_products[df_products['kategorie'] == kategorie].sort_values('preis')print(f"Produkte in '{kategorie}', sortiert (günstig → teuer):")print(gefiltert)print("\nWarum funktioniert die Verkettung?")print("1. df_products[df_products['kategorie'] == kategorie] gibt einen gefilterten DataFrame")print("2. .sort_values('preis') sortiert diesen DataFrame")print("3. Das Ergebnis ist wieder ein DataFrame – wir können weitere Operationen machen!")
```

---## Lösungen zu Aufgabe 3 – SQLite-Datenbanken

### Lösung 3.1: Datenbank aufbauen und Daten einfügen**Was ist die Idee?**- `connect()` erzeugt eine Verbindung (im RAM mit `:memory:`)- `cursor()` ist das Objekt, das SQL-Befehle ausführt- `CREATE TABLE` definiert die Struktur (Spaltennamen und Datentypen)- `executemany()` fügt mehrere Zeilen auf einmal ein

```python
import sqlite3# Verbindung aufbauencon = sqlite3.connect(":memory:")cur = con.cursor()# Tabelle erstellen – mit den gleichen Spalten wie unser DataFrame!cur.execute("""CREATE TABLE produkte (    id INTEGER PRIMARY KEY,    name TEXT NOT NULL,    preis REAL NOT NULL,    kategorie TEXT NOT NULL)""")# Daten einfügenrows = [(p["id"], p["name"], p["preis"], p["kategorie"]) for p in products]cur.executemany(    "INSERT INTO produkte (id, name, preis, kategorie) VALUES (?, ?, ?, ?)",    rows)con.commit()print(f"{len(rows)} Produkte eingefügt.")print("\nWarum .commit()?")print("- Damit werden Änderungen tatsächlich in die Datenbank geschrieben")print("- Ohne .commit() sind Änderungen nur im RAM und gehen verloren!")print("\nWarum Platzhalter '?'?")print("- Sicherer vor SQL-Injection-Angriffen")print("- Bessere Fehlerbehandlung bei Datentypen")
```

### Lösung 3.2: SQL-Abfrage mit Filterung und Sortierung**Was ist die Idee?**- `SELECT` wählt Spalten aus- `WHERE` filtert Zeilen (wie `if` in Python)- `ORDER BY DESC` sortiert absteigend (teuerste zuerst)- `?` sind Platzhalter für Parameter

```python
# SQL-Abfrage: Produkte > 50€, sortiert nach Preis absteigendcur.execute("""SELECT name, preis, kategorie FROM produkte WHERE preis > 50 ORDER BY preis DESC""")ergebnis = cur.fetchall()print("Produkte über 50€ (teuerste zuerst):")for row in ergebnis:    name, preis, kategorie = row    print(f"  {name}: {preis}€ ({kategorie})")print("\nWarum ORDER BY DESC?")print("- ASC (aufsteigend, Standard): 25€, 50€, 100€, 300€")print("- DESC (absteigend): 300€, 100€, 50€, 25€")
```

### Lösung 3.3: Gruppenstatistiken (GROUP BY)**Was ist die Idee?**- `GROUP BY kategorie` fasst Zeilen mit gleicher Kategorie zusammen- `AVG(preis)` berechnet den Durchschnitt pro Gruppe- Das Resultat ist eine kleine Tabelle mit Kategorie und Durchschnitt

```python
# SQL-Abfrage mit Aggregationcur.execute("""SELECT kategorie, AVG(preis) as durchschnitt_preis FROM produkte GROUP BY kategorie ORDER BY durchschnitt_preis DESC""")durchschnitte = cur.fetchall()print("Durchschnittspreise pro Kategorie (teuerste zuerst):")for kategorie, avg_preis in durchschnitte:    print(f"  {kategorie}: {avg_preis:.2f}€")# Zu DataFrame für weitere Analysendf_durchschnitte = pd.DataFrame(durchschnitte, columns=["kategorie", "durchschnitt_preis"])print(f"\nTeuerste Kategorie: {df_durchschnitte.iloc[0]['kategorie']}")print("\nWarum GROUP BY?")print("- Fasst Daten zusammen (Aggregation)")print("- AVG(), COUNT(), SUM(), MIN(), MAX() sind weitere Aggregations-Funktionen")print("- Das Resultat hat weniger Zeilen, aber sagt mehr aus!")
```

## Musterlösung zur Projektaufgabe
Die erwarteten Stammdaten sind drei Kunden (IDs 1–3) und vier Artikel (1001–1004).

```python
import csv, sqlite3
from pathlib import Path
con = sqlite3.connect(":memory:")
con.executescript(Path("data/rechnung_schema.sql").read_text(encoding="utf-8"))
with open("data/kunden.csv", encoding="utf-8", newline="") as f:
    con.executemany("INSERT INTO kunden VALUES (?, ?, ?, ?, ?)", [tuple(r.values()) for r in csv.DictReader(f)])
with open("data/artikel.csv", encoding="utf-8", newline="") as f:
    con.executemany("INSERT INTO artikel VALUES (?, ?, ?, ?)", [(r["artikel_id"], r["bezeichnung"], float(r["einzelpreis"]), float(r["mwst_satz"])) for r in csv.DictReader(f)])
kunde = con.execute("SELECT * FROM kunden WHERE kunden_id = ?", (2,)).fetchone()
artikel = con.execute("SELECT * FROM artikel WHERE artikel_id = ?", (1002,)).fetchone()
if kunde is None or artikel is None:
    raise LookupError("Kunde oder Artikel nicht gefunden")
positionssumme = round(3 * artikel[2], 2)
print(kunde[1], artikel[1], positionssumme, "CHF", artikel[3], "%")
assert positionssumme == 26.70
```
