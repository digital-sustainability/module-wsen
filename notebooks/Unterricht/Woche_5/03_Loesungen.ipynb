## Lösung Aufgabe 1

```python
products = [
    {"id": 1, "name": "Tastatur", "preis": 49.90, "lager": 12},
    {"id": 2, "name": "Maus", "preis": 24.50, "lager": 8},
    {"id": 3, "name": "Monitor", "preis": 219.00, "lager": 5},
    {"id": 4, "name": "Webcam", "preis": 79.00, "lager": 16},
    {"id": 5, "name": "USB-C Hub", "preis": 34.90, "lager": 9},
]
products
```

## Lösung Aufgabe 2

```python
import pandas as pd

df_products = pd.DataFrame(products)
avg_price = df_products["preis"].mean()
low_stock = df_products[df_products["lager"] < 10]

print(f"Durchschnittspreis: {avg_price:.2f}")
low_stock
```

## Lösung Aufgabe 3

```python
import sqlite3

con = sqlite3.connect(":memory:")
cur = con.cursor()

cur.execute("""
CREATE TABLE produkte (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    preis REAL NOT NULL,
    lager INTEGER NOT NULL
)
""")

rows = [(p["id"], p["name"], p["preis"], p["lager"]) for p in products]
cur.executemany("INSERT INTO produkte (id, name, preis, lager) VALUES (?, ?, ?, ?)", rows)
con.commit()

cur.execute("SELECT id, name, preis, lager FROM produkte WHERE preis >= ? ORDER BY preis DESC", (20,))
query_result = cur.fetchall()
query_result
```

## Lösung Aufgabe 4

```python
df_query = pd.DataFrame(query_result, columns=["id", "name", "preis", "lager"])
df_query
```
