## Datenbank öffnen

Um aus Python auf eine SQLite-Datenbank zugreifen zu können, muss die Library **sqlite3** importiert werden. Danach kann mit der Methode **connect()** eine Verbindung zur Datenbank aufgemacht werden. Falls die Datenbank noch nicht existiert, wird diese erstellt.

```python
import sqlite3

con=sqlite3.connect("example.db")
print("The database has been opened successfully!")

#Schliessen der Verbindung
con.close()
```
