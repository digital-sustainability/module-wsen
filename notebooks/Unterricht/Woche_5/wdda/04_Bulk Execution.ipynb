## Mehrfaches Ausführen eines SQL-Befehls

Es ist auch möglich, mit einer Python-Methode den gleichen SQL-Befehl mehrfach auszuführen, aber mit unterschiedlichen Werten pro Ausführung  (sog. _Bulk-Execution_). Zum Beispiel um mehrere Datensätze auf einmal auszuführen, kann zuerst eine Liste mit allen Datensätzen (ein Tupel pro Datensatz) erzeugt werden und dann der Methode **executemany()** übergeben werden.

Beispiel mit einer Liste von Tupeln:

```python
import sqlite3

con = sqlite3.connect('example.db')
cursorObject=con.cursor()

instances = [(10, 'Annika', 44, 'Basel', 5000, 'ja', 'kunst'),
             (11, 'Regula', 24, 'St. Gallen', 4000, 'nein', 'IT'),
             (12, 'Roger', 33, 'Basel', 15000, 'ja', 'wirtschaft'),
             (13, 'Meyrem', 51, 'Bern', 8000, 'ja', 'kunst')
            ]

cursorObject.executemany("INSERT INTO Mitarbeiter VALUES(?,?,?,?,?,?,?)", instances)

print("Daten eingefügt.")

con.commit()
```

Aufgepasst: Die Anzahl Fragezeichen muss gleich der Anzahl Werte in jedem Tupel der Liste sein. Im Beispiel oben entspricht das einem Fragezeichen pro Spalte in der Tabelle.

Beispiel mit Liste von Dictionaries:

```python
employee_1 = {
  "id": 14,
  "name": "Alice",
  "alter": 51,
  "adresse": "Montreux",
  "lohn": 9500,
  "verheiratet": "ja",
  "departement": "IT"
}

employee_2 = {
  "id": 15,
  "name": "Bob",
  "alter": 23,
  "adresse": "Schaffhausen",
  "lohn": 13000,
  "verheiratet": "nein",
  "departement": "Marketing"
}

employee_3 = {
  "id": 16,
  "name": "Claudia",
  "alter": 44,
  "adresse": "Luzern",
  "lohn": 12500,
  "verheiratet": "ja",
  "departement": "Sales"
}

employees = [employee_1, employee_2, employee_3]

cursorObject.executemany("INSERT INTO Mitarbeiter VALUES(:id,:name,:alter,:adresse,:lohn,:verheiratet,:departement)", employees)

print("Daten eingefügt.")

con.commit()
con.close()
```
