## Datenabfrage mittels SELECT

Datenabfragen können ebenfalls via die **execute()**-Methode gemacht werden. Nach der Durchführung können alle (mit **fetchall()**) oder nur einzelne Zeilen (mit **fetchone()** oder **fetchmany()**) des resultierenden Daten abgefragt und ausgegeben werden.

```python
import sqlite3

con=sqlite3.connect("example.db")
cursorObject = con.cursor()

cursorObject.execute("SELECT * FROM Mitarbeiter")
rows=cursorObject.fetchall()

for row in rows:
    print("ID: ", row[0])
    print("Name: ", row[1])
    print("Alter: ", row[2])
    print("Adresse: ", row[3])
    print("Lohn: ", row[4])
    print("Verheiratet: ", row[5])
    print("Department: ", row[6])
    
con.close()
```

Die Ausgabe lässt sich natürlich auch schöner formatieren:


```python
from colorama import Fore
from colorama import Style

print(
    f"{Fore.BLUE}{'id':>4} | {'name':<12} | {'alter':>5} | {'adresse':<30} | \
    {'lohn':>10} | {'verheiratet':<12} | {'departement':<20}")
print("_"*111,  f"{Style.RESET_ALL}")

for row in rows:
    print(f"{row[0]:>4} | {row[1]:<12} | {row[2]:>5} | {row[3]:<30} | \
    {row[4]:>10} | {row[5]:<12} | {row[6]:<20}")

print()
print(f"{len(rows)} rows returned.")
```
