# SQL-Injection: Mitigierung durch parametrisierten Funktionsaufruf

Indem der **execute()**-Methode nicht einfach ein konkatenierter String mitgegeben wird, sondern stattdessen die Variablen als Parameter mitgegeben werden, kann das Problem des vorherigen Codes mitigiert werden.

```python
import sqlite3
from IPython.display import display_markdown

con = sqlite3.connect('example.db')
cursorObject=con.cursor()

name_filter = input("Welche Mitarbeiter sollen zurückgeliefert werden (Wildcards erlaubt): ")
dept_filter = input("Aus welchen Abteilungen (Wildcards erlaubt): ")


# sqlQuery=f"SELECT name, adresse, departement FROM Mitarbeiter WHERE name LIKE '{name_filter}' AND departement LIKE '{dept_filter}'"
sqlQuery="SELECT name, adresse, departement FROM Mitarbeiter WHERE name LIKE ? AND departement LIKE ?"
results=cursorObject.execute(sqlQuery, (name_filter, dept_filter))

rows=results.fetchall()


# Formatierung des Results als Markdown
markdown = '|'
line = '|'

for col in results.description:   
    markdown += col[0] + ' | ' 
    line += '---|'
    
markdown += '\n' + line

for row in rows:
    markdown += f'\n|'
    for i in range(len(results.description)):
        markdown += f' {row[i]} |'

display_markdown(markdown, raw=True)
    
con.close()
```
