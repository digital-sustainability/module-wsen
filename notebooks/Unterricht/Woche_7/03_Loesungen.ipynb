## 1) Sicher rechnen (`calculate`)

```python
def calculate(number1, number2, operator):
    if operator == "+":
        return number1 + number2
    if operator == "-":
        return number1 - number2
    if operator == "*":
        return number1 * number2
    if operator == "/":
        return number1 / number2
    raise ValueError(f"Unbekannter Operator: {operator}")
```

```python
print(calculate(8, 2, "+"))
print(calculate(8, 2, "-"))
print(calculate(8, 2, "*"))
print(calculate(8, 2, "/"))
```

## 2) Zahl robust einlesen (`ask_for_number`)
Die Idee ist einfach: Solange die Eingabe keine Zahl ist, wird die Frage wiederholt.


```python
def ask_for_number(ausgabetext):
    while True:
        eingabe = input(ausgabetext)
        try:
            # Wenn float(eingabe) klappt, haben wir eine gültige Zahl.
            return float(eingabe)
        except ValueError:
            # Bei ungültiger Eingabe bleibt das Programm im Loop und fragt erneut.
            print("Ungültige Eingabe. Bitte eine Zahl eingeben.")
```

## 3) Mini-Taschenrechner (`calculator`)

```python
def calculator():
    number1 = ask_for_number("Erste Zahl: ")
    operator = input("Operator (+, -, *, /): ")
    number2 = ask_for_number("Zweite Zahl: ")

    try:
        result = calculate(number1, number2, operator)
        print("Resultat:", result)
    except ZeroDivisionError:
        print("Fehler: Division durch 0 ist nicht erlaubt.")
    except ValueError as err:
        print("Fehler:", err)
```

## 4) ID-Text parsen mit Validierung

```python
beispiel = """
IDCHE123412345<<<<<<<<<<<<<<<
MUSTER<<ANNA<MARIA<<<<<<<<<<
"""
```

```python
def whitespace_entfernen(text):
    return "".join(text.split())

def id_nummer_lesen(id_text_ohne_ws):
    prefix = "IDCHE"
    if not id_text_ohne_ws.startswith(prefix):
        raise ValueError("Ungültiges ID-Format: Prefix IDCHE fehlt.")

    nummer = id_text_ohne_ws[len(prefix):len(prefix) + 9]
    if len(nummer) != 9 or not nummer.isdigit():
        raise ValueError("Ungültiges ID-Format: ID-Nummer fehlt oder ist nicht numerisch.")

    return nummer
```

```python
beispiel_ohne_ws = whitespace_entfernen(beispiel)
print(id_nummer_lesen(beispiel_ohne_ws))
```

## Musterlösung: Validierung ohne fehlerhafte Rechnung


```python
ERLAUBTE_MWST = (0.0, 2.5, 3.7, 7.7)
def validiere_rechnungsdaten(kunde, artikel, anzahl, einzelpreis, mwst_satz):
    """Zentrale, erwartbare Eingabefehler als ValueError melden."""
    if not isinstance(kunde, dict) or not kunde.get("kunden_id") or not kunde.get("name"):
        raise ValueError("Kunde fehlt oder ist unvollständig")
    if not isinstance(artikel, dict) or not artikel.get("artikel_id") or not artikel.get("bezeichnung"):
        raise ValueError("Artikel fehlt oder ist unvollständig")
    try: menge, preis, satz = float(anzahl), float(einzelpreis), float(mwst_satz)
    except (TypeError, ValueError) as exc: raise ValueError("Menge, Preis und MWST müssen Zahlen sein") from exc
    if menge <= 0: raise ValueError("Menge muss positiv sein")
    if preis < 0: raise ValueError("Preis darf nicht negativ sein")
    if satz not in ERLAUBTE_MWST: raise ValueError("MWST-Satz ist nicht erlaubt")
    return {"anzahl": menge, "einzelpreis": preis, "mwst_satz": satz}

guter_datensatz = validiere_rechnungsdaten({"kunden_id": 1, "name": "Alpen Technik AG"}, {"artikel_id": 1001, "bezeichnung": "USB-C Kabel"}, 2, 12.50, 7.7)
print("Gültig:", guter_datensatz)
for fehler in [(None, {"artikel_id": 1, "bezeichnung": "x"}, 1, 1, 7.7), ({"kunden_id": 1, "name": "x"}, {"artikel_id": 1, "bezeichnung": "x"}, -1, 1, 7.7), ({"kunden_id": 1, "name": "x"}, {"artikel_id": 1, "bezeichnung": "x"}, "zwei", 1, 7.7), ({"kunden_id": 1, "name": "x"}, {"artikel_id": 1, "bezeichnung": "x"}, 1, 1, 8.0)]:
    try: validiere_rechnungsdaten(*fehler)
    except ValueError as exc: print("Erwarteter Fehler:", exc)
```
