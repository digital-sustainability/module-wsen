## 1) Sicher rechnen (`calculate`)

```python
def calculate(number1, number2, operator):
    if operator == "+":
        return number1 + number2
    if operator == "-":
        return number1 - number2
    if operator == "*":
        return number1 * number2
    if operator == "/":
        return number1 / number2
    raise ValueError(f"Unbekannter Operator: {operator}")
```

```python
print(calculate(8, 2, "+"))
print(calculate(8, 2, "-"))
print(calculate(8, 2, "*"))
print(calculate(8, 2, "/"))
```

## 2) Zahl robust einlesen (`ask_for_number`)
Die Idee ist einfach: Solange die Eingabe keine Zahl ist, wird die Frage wiederholt.


```python
def ask_for_number(ausgabetext):
    while True:
        eingabe = input(ausgabetext)
        try:
            # Wenn float(eingabe) klappt, haben wir eine gültige Zahl.
            return float(eingabe)
        except ValueError:
            # Bei ungültiger Eingabe bleibt das Programm im Loop und fragt erneut.
            print("Ungültige Eingabe. Bitte eine Zahl eingeben.")
```

## 3) Projektaufgabe: sichere Rechnungseingabe
Die Lösung prüft einfache Stammdaten direkt und verwendet `try`/`except ValueError` für die Zahleneingaben. So bleibt die Datenstruktur in SW7 bewusst bei einzelnen Werten und Tupeln.


```python
ERLAUBTE_MWST = (0.0, 2.5, 3.7, 7.7)

def validiere_rechnungsdaten(kunde_id, kunde_name, artikel_nummer, artikel_name, menge_input, einzelpreis_input, mwst_input):
    if not kunde_id or not kunde_name:
        raise ValueError("Kundendaten sind unvollständig.")
    if not artikel_nummer or not artikel_name:
        raise ValueError("Artikeldaten sind unvollständig.")

    try:
        menge = int(menge_input)
        einzelpreis = float(einzelpreis_input)
        mwst_satz = float(mwst_input)
    except ValueError as exc:
        raise ValueError("Menge, Preis und MWST müssen Zahlen sein.") from exc

    if menge <= 0:
        raise ValueError("Menge muss positiv sein.")
    if einzelpreis < 0:
        raise ValueError("Preis darf nicht negativ sein.")
    if mwst_satz not in ERLAUBTE_MWST:
        raise ValueError("MWST-Satz ist nicht erlaubt.")

    kunde = (kunde_id, kunde_name)
    artikel = (artikel_nummer, artikel_name)
    position = (menge, einzelpreis, mwst_satz)
    return kunde, artikel, position

guter_datensatz = validiere_rechnungsdaten("K100", "Alpen Technik AG", "A200", "USB-C Kabel", "2", "12.5", "7.7")
print("Gültig:", guter_datensatz)

fehlerhafte_datensaetze = [
    ("", "Alpen Technik AG", "A200", "USB-C Kabel", "2", "12.5", "7.7"),
    ("K100", "Alpen Technik AG", "A200", "USB-C Kabel", "0", "12.5", "7.7"),
    ("K100", "Alpen Technik AG", "A200", "USB-C Kabel", "zwei", "12.5", "7.7"),
    ("K100", "Alpen Technik AG", "A200", "USB-C Kabel", "2", "12.5", "8.0"),
]

for daten in fehlerhafte_datensaetze:
    try:
        validiere_rechnungsdaten(*daten)
    except ValueError as exc:
        print("Erwarteter Fehler:", exc)
```

## 4) Mini-Taschenrechner (`calculator`)


```python
def calculator():
    number1 = ask_for_number("Erste Zahl: ")
    operator = input("Operator (+, -, *, /): ")
    number2 = ask_for_number("Zweite Zahl: ")

    try:
        result = calculate(number1, number2, operator)
        print("Resultat:", result)
    except ZeroDivisionError:
        print("Fehler: Division durch 0 ist nicht erlaubt.")
    except ValueError as err:
        print("Fehler:", err)
```

## 5) ID-Text parsen mit Validierung


```python
beispiel = """
IDCHE123412345<<<<<<<<<<<<<<<
MUSTER<<ANNA<MARIA<<<<<<<<<<
"""
```

```python
def whitespace_entfernen(text):
    return "".join(text.split())

def id_nummer_lesen(id_text_ohne_ws):
    prefix = "IDCHE"
    if not id_text_ohne_ws.startswith(prefix):
        raise ValueError("Ungültiges ID-Format: Prefix IDCHE fehlt.")

    nummer = id_text_ohne_ws[len(prefix):len(prefix) + 9]
    if len(nummer) != 9 or not nummer.isdigit():
        raise ValueError("Ungültiges ID-Format: ID-Nummer fehlt oder ist nicht numerisch.")

    return nummer
```

```python
beispiel_ohne_ws = whitespace_entfernen(beispiel)
print(id_nummer_lesen(beispiel_ohne_ws))
```
