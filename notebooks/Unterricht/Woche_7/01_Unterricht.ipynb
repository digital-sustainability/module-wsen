## 1. Fehlerarten verstehen

### 1.1 Was ist ein Fehler?
Ein Fehler ist ein Hinweis von Python, dass etwas im Programm nicht so abläuft wie erwartet. Das ist kein Grund zur Panik – es ist eine Chance, den Code besser zu verstehen und robuster zu machen.

### 1.2 Syntaxfehler, Laufzeitfehler und logische Fehler
- **Syntaxfehler:** Der Code ist falsch geschrieben. Python kann ihn nicht lesen, deshalb startet das Programm nicht.
- **Laufzeitfehler:** Der Code ist syntaktisch korrekt, aber beim Ausführen passiert etwas Unerwartetes, zum Beispiel eine ungültige Eingabe.
- **Logischer Fehler:** Der Code läuft durch, aber das Ergebnis ist falsch. Es gibt dann keine Fehlermeldung – die Logik selbst ist das Problem.

Beispiele:
- Syntaxfehler: `print("Hallo"`
- Laufzeitfehler: `float("abc")`
- Logischer Fehler: Ein Programm rechnet `2 + 2` falsch aus, obwohl es keine Exception erzeugt.


```python
text = "12.5"
zahl = float(text)
print(zahl)
```

```python
# Laufzeitfehler (ValueError), weil "abc" keine Zahl ist
float("abc")
```

### 1.3 Warum brauchen wir Fehlerbehandlung?
Wenn wir Benutzereingaben verarbeiten, kann etwas Unerwartetes passieren. Ohne Fehlerbehandlung würde das Programm an dieser Stelle abbrechen. Mit `try` und `except` können wir auf solche Fälle reagieren und dem Nutzer eine verständliche Rückmeldung geben.

Beispiel:
- Ohne Schutz: `zahl = int(input("Bitte eine Zahl eingeben: "))`
- Mit Schutz: `try/except` fragt erneut nach oder gibt eine hilfreiche Nachricht aus.


```python
while True:
    try:
        eingabe = input("Bitte eine Zahl eingeben: ")
        zahl = float(eingabe)
        print("Eingabe erfolgreich:", zahl)
        break
    except ValueError:
        print("Das war keine Zahl. Bitte versuche es erneut.")
```

### 1.4 `try` und `except`
Mit `try` markieren wir Code, der fehlschlagen kann. Wenn dort ein Fehler entsteht, springt Python direkt in den `except`-Block. So bleibt das Programm kontrollierbar.


```python
eingabe = "abc"

try:
    zahl = float(eingabe)
    print("Eingabe als Zahl:", zahl)
except ValueError:
    print("Ungültige Eingabe")
```

Wichtig: In der Praxis sollten wir möglichst **konkrete Fehlertypen** (z. B. `ValueError`) abfangen statt eines allgemeinen `except`.

```python
eingabe = "abc"

try:
    zahl = float(eingabe)
    print("Eingabe als Zahl:", zahl)
except ValueError:
    print("Bitte eine Zahl eingeben.")
```

## Anwendung im Projekt: Zahlenfelder sicher prüfen
Bei einer Rechnungsposition kommen Menge, Einzelpreis und MWST oft zuerst als Text an. Statt den Typ separat zu prüfen, wandeln wir die Werte direkt mit `try`/`except` um und melden verständliche Fehler.


```python
ERLAUBTE_MWST = (0.0, 2.5, 3.7, 7.7)

def position_pruefen(menge_input, einzelpreis_input, mwst_input):
    try:
        menge = int(menge_input)
        einzelpreis = float(einzelpreis_input)
        mwst_satz = float(mwst_input)
    except ValueError as exc:
        raise ValueError("Menge, Preis und MWST müssen als Zahl eingegeben werden.") from exc

    if menge <= 0:
        raise ValueError("Menge muss positiv sein.")
    if einzelpreis < 0:
        raise ValueError("Preis darf nicht negativ sein.")
    if mwst_satz not in ERLAUBTE_MWST:
        raise ValueError("MWST-Satz ist nicht erlaubt.")

    return menge, einzelpreis, mwst_satz

print(position_pruefen("2", "12.5", "7.7"))
try:
    position_pruefen("zwei", "12.5", "7.7")
except ValueError as exc:
    print("Erwarteter Fehler:", exc)
```

### 1.5 Mehrere Fehlertypen gezielt behandeln
Im folgenden Beispiel können zwei verschiedene Fehler auftreten: `ValueError` (ungültige Zahl) und `ZeroDivisionError` (Division durch 0).


```python
def sichere_division(text, nenner):
    try:
        zaehler = float(text)
        return zaehler / nenner
    except ValueError:
        print("Fehler: Der Zähler ist keine Zahl.")
    except ZeroDivisionError:
        print("Fehler: Division durch 0 ist nicht erlaubt.")

print(sichere_division("12", 3))
print(sichere_division("abc", 3))
print(sichere_division("12", 0))
```

### 1.6 `else` und `finally`
- `else` wird nur ausgeführt, wenn im `try`-Block **kein** Fehler auftritt.
- `finally` wird **immer** ausgeführt, unabhängig davon, ob ein Fehler auftrat.

Das ist nützlich, wenn wir z. B. am Ende immer eine Abschlussmeldung oder Aufräumarbeiten brauchen.


```python
def text_zu_float(text):
    try:
        wert = float(text)
    except ValueError:
        print("Konvertierung fehlgeschlagen")
        return None
    else:
        print("Konvertierung erfolgreich")
        return wert
    finally:
        print("Konvertierungsversuch abgeschlossen")

text_zu_float("3.5")
text_zu_float("hallo")
```

### 1.7 Robuste Eingabefunktion bauen
Ein typisches Muster in Programmen: Wir fragen so lange nach, bis eine gültige Eingabe vorliegt. Dazu kombinieren wir `while True` mit `try`/`except`.


```python
def ask_for_number(ausgabetext):
    while True:
        eingabe = input(ausgabetext)
        try:
            return float(eingabe)
        except ValueError:
            print("Das war keine Zahl. Bitte versuche es erneut.")
```

Diese Funktion ist ein gutes Praxisbeispiel für die Wiederverwendung aus Woche 6: Einmal sauber definiert, kann sie in vielen Programmen direkt genutzt werden.

## Anwendung im Projekt: Rechnungsdaten gemeinsam validieren
Auch Kunden- und Artikeldaten prüfen wir mit einfachen Bedingungen. Für diesen einzelnen Datensatz verwenden wir weiterhin einzelne Werte und Tupel; die Zahlenprüfung aus dem vorherigen Block wird direkt wiederverwendet.


```python
def validiere_rechnungsdaten(kunde_id, kunde_name, artikel_nummer, artikel_name, menge_input, einzelpreis_input, mwst_input):
    if not kunde_id or not kunde_name:
        raise ValueError("Kundendaten sind unvollständig.")
    if not artikel_nummer or not artikel_name:
        raise ValueError("Artikeldaten sind unvollständig.")

    menge, einzelpreis, mwst_satz = position_pruefen(menge_input, einzelpreis_input, mwst_input)
    kunde = (kunde_id, kunde_name)
    artikel = (artikel_nummer, artikel_name)
    position = (menge, einzelpreis, mwst_satz)
    return kunde, artikel, position

print(validiere_rechnungsdaten("K100", "Alpen Technik AG", "A200", "USB-C Kabel", "2", "12.5", "7.7"))
try:
    validiere_rechnungsdaten("", "Alpen Technik AG", "A200", "USB-C Kabel", "2", "12.5", "7.7")
except ValueError as exc:
    print("Erwarteter Fehler:", exc)
```

## 2. Listen: mehrere Werte gemeinsam speichern

Eine Liste speichert mehrere Werte in einer festen Reihenfolge. In Python schreiben wir Listen mit eckigen Klammern. Der allgemeinere Begriff **Array** bezeichnet ebenfalls eine geordnete Sammlung von Werten. Für diese Einführung verwenden wir die eingebaute Python-Liste (`list`); zusätzliche Bibliotheken oder NumPy-Arrays brauchen wir nicht.


```python
artikel = ["Bleistift", "Heft", "Lineal"]
print(artikel)
print("Erster Artikel:", artikel[0])
print("Anzahl Artikel:", len(artikel))
```

### 2.1 Mit Index auf Elemente zugreifen
Der Index ist die Position eines Elements. In Python beginnt das Zählen bei `0`: `artikel[0]` ist das erste und `artikel[1]` das zweite Element. Ein Index ausserhalb der Liste, zum Beispiel `artikel[3]`, löst einen `IndexError` aus. Mit `len(artikel)` erhältst du die Anzahl der Elemente.


```python
artikel[1] = "Notizheft"
artikel.append("Radiergummi")
print(artikel)
```

Listen sind veränderbar: Mit `liste[index] = wert` ersetzen wir ein Element und mit `append(wert)` hängen wir ein Element am Ende an.

### 2.2 Eine Liste mit `for` durchlaufen
Mit einer `for`-Schleife bearbeiten wir jedes Element der Reihe nach. Das verbindet die Listen mit den Schleifen aus Woche 4.


```python
preise = [2.50, 4.00, 1.75]
total = 0

for preis in preise:
    total = total + preis

print("Einzelpreise:")
for preis in preise:
    print(preis)
print("Total:", total)
```

Du kannst mit einer Schleife also nicht nur alle Werte ausgeben, sondern auch eine Summe aufbauen. Hier steht jeder Listenwert für einen Artikelpreis; später begegnen uns Listen auch als Beziehungen zwischen Objekten.
