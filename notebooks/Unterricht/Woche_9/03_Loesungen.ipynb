## Aufgabe 1 (geführt): Tuple-Grundlagen**Erklärung:**- Ein Tuple mit `(Wert, Wert, Wert)` speichert mehrere Werte in fester Reihenfolge.- Mit `[0]`, `[1]`, `[2]` greifst du auf die einzelnen Werte zu (Index startet bei 0).- "Unpacking" ist eine Kurzschreibweise: statt `x = punkt[0]; y = punkt[1]` schreibst du einfach `x, y = punkt`.

```python
person = ("Mira", 19, "Informatik")# Mit Index auslesenname = person[0]       # "Mira"alter = person[1]      # 19fach = person[2]       # "Informatik"# Tuple Unpackingpunkt = (4, 9)x, y = punkt  # x=4, y=9print("Person:", name, ",", alter, "Jahre alt,", "Fach:", fach)print("Koordinaten: x =", x, ", y =", y)
```

## Aufgabe 2 (geführt): Dictionary-Grundlagen**Erklärung:**- Ein Dictionary `{"Key": Wert}` speichert Paare aus Schlüssel und Wert.- Mit `dict["Key"]` greifst du auf den Wert zu (nicht mit Index wie bei Listen!).- Mit `dict["Key"] = neuer_Wert` änderst du einen Wert oder fügst einen neuen Eintrag hinzu.- Mit `.items()` iterierst du über alle Key-Value-Paare auf einmal.

```python
noten = {"Aylin": 5.0, "Noah": 4.5}# Note auslesenprint("Ausgelesene Note:", noten["Aylin"])  # 5.0# Vorhandenen Eintrag ändernnoten["Noah"] = 5.0# Neuen Eintrag hinzufügennoten["Lina"] = 5.5# Alle Einträge mit .items() iterierenprint("Alle Noten:")for schueler, note in noten.items():    print(schueler, "->", note)# Warum .items()? Damit kriegst du pro Schleife zwei Werte:# den Key ("Aylin") und den Value (5.0)# Wenn du nur dict.keys() nimmst, kriegst du nur "Aylin", "Noah", "Lina"
```

## Aufgabe 3 (teilweise gefuehrt): Dictionary umkehren


```python
def dictionary_umkehren(dictionary):
    neuer_dictionary = {}
    for key, value in dictionary.items():
        # Value wird zum neuen Key, Key wird zum neuen Value
        neuer_dictionary[value] = key
    return neuer_dictionary


print(dictionary_umkehren({"a": 1, "b": 2}))
```

## Aufgabe 5 (Transfer): Wort-Frequenzen


```python
def word_freq(text):
    haeufigkeiten = {}

    for wort in text.split():
        if wort in haeufigkeiten:
            haeufigkeiten[wort] += 1
        else:
            haeufigkeiten[wort] = 1

    return haeufigkeiten


print(word_freq("This is a simple example. This is a test example."))
```
