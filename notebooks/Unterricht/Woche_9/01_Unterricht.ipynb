## 1. Tuples

### 1.1 Was ist ein Tuple?

Ein Tuple ist eine **geordnete Sammlung** von Werten.
Die Reihenfolge bleibt erhalten, und die Anzahl Eintraege ist oft von Anfang an klar.

Typische Beispiele:
- Koordinate: `(x, y)`
- RGB-Farbe: `(rot, gruen, blau)`
- Stundenplaneintrag: `(Modul, Raum, Startzeit)`

> **Merkregel:** Tuple = geordnet + unveraenderlich.


````python
koordinate = (47.3769, 8.5417)
farbe_rgb = (255, 128, 0)
lektion = ("Datenbanken", "B201", "08:15")

print("Koordinate:", koordinate)
print("RGB:", farbe_rgb)
print("Stundenplan:", lektion)
````

### 1.2 Erstellen und Elemente abrufen

Wie bei Listen greifen wir mit Indexen zu.
- erster Wert: Index `0`
- zweiter Wert: Index `1`
- dritter Wert: Index `2`

Gerade bei festen Datensaetzen ist das sehr praktisch.


````python
student = ("Mira", 19, "Informatik")

name = student[0]
alter = student[1]
fach = student[2]

print("Name:", name)
print("Alter:", alter)
print("Fach:", fach)
````

### 1.3 Tuple Unpacking

Beim Unpacking verteilen wir Tuple-Werte direkt auf Variablen.
Das macht den Code kuerzer und oft besser lesbar.


````python
punkt = (3, 7)
x, y = punkt

print("x =", x)
print("y =", y)

rgb = (34, 139, 34)
r, g, b = rgb
print("r, g, b =", r, g, b)
````

### 1.4 Unveraenderlichkeit (immutability)

Tuple-Inhalte koennen nicht direkt geaendert werden.
Das hilft bei Daten, die stabil bleiben sollen, z. B.:
- feste Kennungen
- Koordinaten
- Zeitstempel-Paare

Vorteil: Du vermeidest versehentliche Aenderungen im spaeteren Programmablauf.


````python
fixe_id = (2026, "SW09", 17)
print("Fixe ID:", fixe_id)

# fixe_id[0] = 2027  # TypeError: Tuple ist unveraenderlich

# Wenn du etwas aendern willst, erzeugst du ein neues Tuple:
neue_id = (2027, fixe_id[1], fixe_id[2])
print("Neue ID:", neue_id)
````

### 1.5 Wann Tuple, wann Liste?

| Frage | Tuple | Liste |
|---|---|---|
| Reihenfolge wichtig? | Ja | Ja |
| Inhalt oft aendern? | Nein | Ja |
| Typischer Einsatz | Koordinaten, feste Datensaetze | To-do-Liste, dynamische Sammlungen |

**Praxisentscheidung:**
- `todo = ["lernen", "essen", "sport"]` -> Liste, weil sich Eintraege oft aendern.
- `pruefung = ("M123", "Raum B201", "09:00")` -> Tuple, weil die Struktur fix ist.

> **Merkregel:** Wenn die Struktur stabil ist, starte mit Tuple.

## 2. Dictionaries

Dictionaries speichern Paare aus `key -> value`.
Du kannst damit sehr schnell etwas "nachschlagen", z. B.:
- `benutzername -> passwort`
- `produkt_nr -> preis`
- `modulcode -> modulname`

Ein Dictionary ist ideal, wenn du ueber einen eindeutigen Schluessel auf Daten zugreifen willst.


---
### 2.0 Dictionary verstehen: Das Telefonbuch-Analoge

Stellt euch ein Telefonbuch vor:
- **Klassisches Telefonbuch:** Name → Telefonnummer (genauso funktioniert ein Dictionary!)
- **Schlag einen Namen nach** → du findest sofort die Nummer
- **Füge einen Eintrag hinzu** → Seite öffnen, neuen Namen schreiben, Nummer daneben
- **Lösche einen Eintrag** → Streichst den Namen durch

In Python ist die Syntax dasselbe:
- **Name nachschlagen:** `telefonbuch["Anna"]` → gibt die Nummer zurück
- **Eintrag hinzufügen:** `telefonbuch["Bob"] = "555-1234"` → neue Zeile hinzufügen
- **Eintrag ändern:** `telefonbuch["Anna"] = "555-5678"` → alte Nummer überschreiben
- **Eintrag löschen:** `del telefonbuch["Anna"]` → Eintrag entfernen

**Das macht Dictionaries so mächtig:**
Du brauchst nicht die 0., 1., 2., 3. Person zu zählen (wie bei Listen mit Index).
Du nennst den Namen und bekommst sofort die Information!


---
### 2.0b Listen vs. Dictionaries: Wann welches?

| Frage | Liste | Dictionary |
|-------|-------|-----------|
| Ich kenne die Position (Index)? | Ja: `liste[2]` | Nein: `dict[name]` |
| Ich brauche schnelle Suche nach Inhalt? | Nein (dauert lang) | Ja (sehr schnell) |
| Die Reihenfolge ist wichtig? | Ja | Nein (aber seit Python 3.7 erhalten!) |
| Beispiel | Messwerte: `[12.5, 13.1, 12.9]` | Namen → Alter: `{"Anna": 41, "Bob": 38}` |

**Praxisbeispiel:**
```python
# LISTEN: Wenn die Position zählt
messungen = [12.5, 13.1, 12.9]
erste_messung = messungen[0]

# DICTIONARIES: Wenn die Bezeichnung zählt
alter = {"Anna": 41, "Bob": 38}
annas_alter = alter["Anna"]
```



### 2.1 Definition, Abfrage und Veraenderung

Eintrag lesen: `dict[key]`
Eintrag aendern: `dict[key] = neuer_wert`
Eintrag hinzufuegen: `dict[neuer_key] = wert`


````python
alter = {"Anna": 41, "Marta": 23}

print("Alter von Anna:", alter["Anna"])

alter["Anna"] = 42
alter["Bettina"] = 53
print("Nach Aenderung:", alter)

preise = {"Maus": 25, "Tastatur": 70}
preise["Maus"] = 27
preise["Headset"] = 49
print("Preisliste:", preise)
````

### 2.2 Was darf in ein Dictionary?

- **Values**: fast alles (Zahlen, Strings, Listen, sogar weitere Dictionaries).
- **Keys**: muessen hashable sein (z. B. String, Zahl, Tuple).
- Listen sind als Key nicht erlaubt, weil sie veraenderbar sind.

Warum ist das wichtig?
Python braucht stabile Keys, damit Eintraege intern korrekt wiedergefunden werden.


````python
rezepte = {
    "Risotto": ["Reis", "Parmesan"],
    "Suppe": ["Kuerbis", "Ingwer"]
}

woerterbuch = {
    ("de", "hallo"): "hello",
    ("de", "tschuess"): "bye"
}

print("Zutaten Risotto:", rezepte["Risotto"])
print("Uebersetzung:", woerterbuch[("de", "hallo")])
````

### 2.3 Elemente loeschen (`del`, `pop`, `clear`)

- `del d[key]` -> loescht ohne Rueckgabe
- `d.pop(key)` -> loescht und liefert den alten Wert
- `d.clear()` -> leert das ganze Dictionary


````python
scores = {"A": 10, "B": 20, "C": 30}
print("Start:", scores)

del scores["A"]
print("Nach del:", scores)

wert = scores.pop("B")
print("pop-Rueckgabe:", wert)
print("Nach pop:", scores)

scores.clear()
print("Nach clear:", scores)
````

### 2.4 Wichtige Methoden (`update`, `get`, `keys`, `values`, `items`)

**`get` vs. eckige Klammern:**
- `d["x"]` wirft Fehler, wenn `"x"` fehlt.
- `d.get("x")` gibt `None` (oder Default-Wert) zurueck.

Das ist in realen Programmen wichtig, wenn Eingaben unvollstaendig sein koennen.


````python
basis = {"x": 1, "y": 2}
neu = {"z": 3}
basis.update(neu)

print("Nach update:", basis)
print("get('x'):", basis.get("x"))
print("get('nicht_da'):", basis.get("nicht_da"))
print("get('nicht_da', 0):", basis.get("nicht_da", 0))
print("keys:", list(basis.keys()))
print("values:", list(basis.values()))
print("items:", list(basis.items()))
````

### 2.5 Ueber Dictionaries iterieren

Mit `.items()` bekommst du pro Durchlauf ein Tuple `(key, value)`.
Das ist die haeufigste Form, wenn du beide Werte gleichzeitig brauchst.


````python
lager = {"Maus": 12, "Tastatur": 8, "Headset": 5}

for artikel, bestand in lager.items():
    print(artikel, "->", bestand)
````

````python
warenkorb = {"Maus": 2, "Tastatur": 1, "USB-Kabel": 3}
preise = {"Maus": 27, "Tastatur": 70, "USB-Kabel": 9}

gesamt = 0
for artikel, menge in warenkorb.items():
    position = menge * preise[artikel]
    gesamt += position
    print(f"{artikel}: {menge} x {preise[artikel]} = {position}")

print("Gesamtpreis:", gesamt)
````

## 3b. Entscheidungshilfe: Welche Datenstruktur für welche Aufgabe?

**Die 3 Datenstrukturen im Überblick:**

### 1. **Liste** — für geordnete, veränderliche Sammlungen
```python
# Wenn du: eine dynamische Sammlung brauchst, die sich ändert
messungen = [12.5, 13.1, 12.9]
messungen.append(13.4)  # neue Messung hinzufügen

# Wenn du: über alle Werte nacheinander iterierst
for messung in messungen:
    print(messung)
```
**Beispiele:** To-do-Liste, Messergebnisse, Einkaufsliste, Chat-Nachrichten

### 2. **Tuple** — für geordnete, unveränderliche Sammlungen
```python
# Wenn du: Daten haben willst, die sich NICHT ändern dürfen
gps = (47.3769, 8.5417)  # Koordinaten sollten nicht versehentlich geändert werden
farbe = (255, 128, 0)    # RGB-Wert ist definiert

# Tuples als Dictionary-Keys
filme = {
    ("2024", "Drama"): ["Film A", "Film B"],
    ("2024", "Action"): ["Film C"]
}
```
**Beispiele:** GPS-Koordinaten, Geburtsdatum, Farbwerte, Eindeutige IDs, Dictionary-Keys

### 3. **Dictionary** — für schnelle Abfragen mit Schlüssel
```python
# Wenn du: Daten über einen NAMEN/SCHLÜSSEL abrufen willst
alter = {"Anna": 41, "Bob": 38, "Carol": 35}
print(alter["Anna"])  # sehr schnell! (nicht erst durch die Liste gehen)

# Wenn du: komplexe Strukturen mit Bezeichnungen speichern willst
student = {
    "name": "Mira",
    "alter": 19,
    "fach": "Informatik",
    "noten": [5.0, 4.5, 5.5]  # auch komplexe Werte möglich
}
```
**Beispiele:** Benutzerprofile, Preise, Telefonbuch, Konfigurationen, Login-Systeme

### Schnelle Entscheidungsregel:
1. **Positioniert?** (0., 1., 2., ...) → **Liste**
2. **Unveränderlich?** → **Tuple**
3. **Nachgeschlagen nach Name/Schlüssel?** → **Dictionary**

### Praxisszenarien:
| Szenario | Struktur | Grund |
|----------|----------|-------|
| Einkaufsliste mit Dingen, die ich noch kaufe | Liste | Veränderlich: Items abhaken, neue hinzufügen |
| Geburtstag speichern (1999, 5, 15) | Tuple | Fest, unveränderlich |
| Preise nachschlagen: "Apfel" → 1.50 € | Dictionary | Schnelle Abfrage nach Name |
| Messwerte von einem Sensor | Liste | Dynamisch: neue Messung alle 10 Sekunden |
| Benutzer-Login speichern: username → passwort | Dictionary | Schnelle Authentifizierung |
| Koordinaten einer Route speichern | Liste von Tuples | Viele Positionen, jede ist fest |
